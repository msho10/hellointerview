### DESCRIPTION (inspired by Leetcode.com)

You are given a chessboard of infinite size where the coordinates of each cell are defined by integer pairs (x, y). The knight piece moves in an L-shape, either two squares horizontally and one square vertically, or two squares vertically and one square horizontally.

Write a function to determine the minimum number of moves required for the knight to move from the starting position (0, 0) to the target position (x, y). Assume that it is always possible to reach the target position, and that x and y are both integers in the range [-200, 200]

Example 1:

Input:
```
x = 1
y = 2
```
Output: 1

Explanation: The knight can move from (0, 0) to (1, 2) in one move.

Example 2:

Input
```
x = 4
y = 4
```
Output: 4

Explanation: The knight can move from (0, 0) to (4, 4) in four moves ( [0, 0] -> [2, 1] -> [4, 2] -> [6, 3] -> [4, 4] )

````python
from collections import deque

class Solution:
    def minimumKnightMoves(self, x: int, y: int) -> int:
        directions = [(1,2),(2,1),(-1,-2),(-2,-1),(-1,2),(-2,1),(1,-2),(2,-1)]
        queue = deque([(0,0)])
        visited = set((0,0))
        moves = -1

        while queue:
            moves += 1
            level_len = len(queue)
            for _ in range(level_len):
                x_, y_ = queue.popleft()
                if x_ == x and y_ == y:
                    return moves

                for dx, dy in directions:
                    nx = x_ + dx
                    ny = y_ + dy
                    if -200 <= nx <= 200 and -200 <= ny <= 200 and (nx, ny) not in visited:
                        visited.add((nx, ny))
                        queue.append((nx,ny)) 

        return moves            
````

### Feedback

Your BFS is correct and passes the tests. Because BFS processes positions level by level, moves correctly represents the distance, and checking the target when dequeuing is valid. visited = set((0,0)) works here because (0, 0) is iterable, though it is less explicit and fragile; prefer visited = {(0, 0)}. 

The main issue is that your implementation unnecessarily restricts the search to [-200, 200]; the problem describes an infinite board, so a general solution should not impose that boundary (the bounded constraints only apply to the target). 

Also ensure deque is imported with from collections import deque. 

Returning moves after the loop is unreachable under the stated assumptions, but return -1 better communicates the fallback.

````python
from typing import Callable

class Input:
    def __init__(self, x: int, y: int):
        self.x = x
        self.y = y
        
class Test:  
    def __init__(self, input: Input, result: int):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[int, int], int]):
    for test in tests:
        result = func(test.input.x, test.input.y)
        if result == test.expected_result:
            print(f"Test passed for ({test.input.x},{test.input.y})")
        else:
            print(f"Test failed for ({test.input.x},{test.input.y}). Expected: {test.expected_result}, Actual: {result}")
````

````python
tests = [
    Test(Input(0,0),0),
    Test(Input(1,2),1),
    Test(Input(4,4),4),
    Test(Input(-1,-2),1),
    Test(Input(-4,-4),4),
]

run_tests(tests, Solution().minimumKnightMoves)
````
