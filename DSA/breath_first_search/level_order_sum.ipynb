Given the root of a binary tree, return the sum of the nodes at each level. The output should be a list containing the sum of the nodes at each level.

Example 1:

Input: [1, 3, 4, null, 2, 7, null, 8]

Output: [1, 7, 9, 8]

Example 2:

Input: [1, 2, 5, 3, null, null, null, null, 4]

Output: [1, 7, 3, 4]

```python
from collections import deque
class TreeNode:
    def __init__(self, val: int, left: 'TreeNode' = None, right: 'TreeNode' = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def level_order_sum(self, root: TreeNode) -> list[int]:
        if root is None:
            return []
        
        result = []
        queue = deque([root])

        while queue:
            level_len = len(queue)
            level_sum = 0

            for i in range(level_len):
                node = queue.popleft()
                level_sum += node.val
                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)

            result.append(level_sum)

        return result
```

### Feedback

Your solution is correct and efficiently performs level-order traversal. Capturing level_len before processing each level ensures children are queued for the next level, so sums are grouped properly. It handles None and negative values correctly. Complexity is O(n) time and O(w) space, where w is the maximum tree width.

```python
from typing import Callable

class Input:
    def __init__(self, nums: list[int]):
        self.nums = nums
        
class Test:  
    def __init__(self, input: Input, result: list[int]):
        self.input = input
        self.expected_result = result
        self.root = self._createNodefromList(input.nums, 0)

    def _createNodefromList(self, nums: list[int], index: int) -> TreeNode:
        if index >= len(nums) or nums[index] is None:
            return None
        newNode = TreeNode(nums[index])
        newNode.left = self._createNodefromList(nums, 2*index+1)
        newNode.right = self._createNodefromList(nums, 2*index+2)
        return newNode

def run_tests(tests: list[Test], func: Callable[[TreeNode], list[int]]):
    for test in tests:
        result = func(test.root)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums}")
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
```
