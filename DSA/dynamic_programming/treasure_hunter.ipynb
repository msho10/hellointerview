### AI prompt

Implement a solution for the following problem using DP bottom-up approach. The solution should use an array of n+1 size where n is length of the input list. 
The state for dp[i] is the max amount of treasures for previous i houses. The final result is in dp[n].

Problem:
You're a treasure hunter in a neighborhood where houses are arranged in a row, and each house contains a different amount of treasure. Your goal is to collect as much treasure as possible, but there's a catch: if you collect treasure from two adjacent houses, it triggers a neighborhood-wide alert, ending your hunt immediately.
Given an array treasure of non-negative integers, where each integer represents the amount of treasure in a house, write a function to return the maximum amount of treasure you can collect without triggering any alarms.

Example 1:

Input: 
treasure = : [3, 1, 4, 1, 5]  

Result = 12

```python
def max_treasure(treasure: list[int]) -> int:
    n = len(treasure)
    if n == 0:
        print("DP Array: [0]")
        return 0
    if n == 1:
        print(f"DP Array: [0, {treasure[0]}]")
        return treasure[0]

    # Initialize dp array of size n + 1
    dp = [0] * (n + 1)
    
    # Base cases
    dp[0] = 0
    dp[1] = treasure[0]
    
    # Fill the dp array bottom-up
    for i in range(2, n + 1):
        dp[i] = max(dp[i - 1], dp[i - 2] + treasure[i - 1])
        
    # Print the full dp array of size n + 1
    print(f"DP Array (size {n + 1}): {dp}")
    return dp[n]

# Example test
treasure = [3, 1, 4, 1, 5]
print(f"Result = {max_treasure(treasure)}")
```

Optimize the solution to reduce the space complexity from O(n) to O(1)

```python
def max_treasure_optimized(treasure: list[int]) -> int:
    n = len(treasure)
    if n == 0:
        return 0
    if n == 1:
        return treasure[0]

    # prev2 represents dp[i-2] (starts at dp[0] = 0)
    # prev1 represents dp[i-1] (starts at dp[1] = treasure[0])
    prev2 = 0
    prev1 = treasure[0]
    
    # Iterate through houses starting from index 1 (house 2)
    for i in range(1, n):
        current = max(prev1, prev2 + treasure[i])
        prev2 = prev1
        prev1 = current
        
    return prev1

# Example test
treasure = [3, 1, 4, 1, 5]
print(f"Result = {max_treasure_optimized(treasure)}")  # Output: 12
```
