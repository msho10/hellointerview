Given a string s, split it into segments where each segment is a palindrome (reads the same forwards and backwards).

Return all possible ways to partition the string into palindromic segments.

Constraints:

1 <= s.length <= 16
s contains only lowercase English letters
Example 1:

Input:
s = "noon"

Output:
[["n","o","o","n"], ["n","oo","n"], ["noon"]]

Explanation:

["n","o","o","n"] - All single characters are palindromes
["n","oo","n"] - "oo" is a palindrome in the middle
["noon"] - The entire word is a palindrome

Example 2:

Input:
s = "civic"

Output:

[["c","i","v","i","c"], ["c","ivi","c"], ["civic"]]

```python
from typing import List

class Solution:
    def partition(self, s: str) -> List[List[str]]:
        def isPalindrome(word: str, start: int, end: int) -> bool:
            while start<end:
                if word[start] != word[end]:
                    return False
                start += 1
                end -= 1
            return True

        def dfs(word: str, start: int):
            if start == len(word):
                result.append(path[:])
                return
            
            for end in range(start, len(word)):
                if isPalindrome(word, start, end):
                    path.append(word[start:end+1])
                    dfs(word, end+1)
                    path.pop()

        result, path = [], []    
        dfs(s, 0)
        return result
```

```python
from typing import Callable

class Test:  
    def __init__(self, s: str, result: List[List[str]]):
        self.s = s
        self.expected_result = result

    def verify(self, actual_result: List[List[str]]) -> False:
        verify_result = True

        if self.expected_result is None:
            return actual_result is None
         
        for item in self.expected_result:
            if item not in actual_result:
                print(f"missing item: {item}")
                verify_result = False

        # in case actual_result contains more:
        if len(self.expected_result) < len(actual_result):
            for item in actual_result:
                if item not in self.expected_result:
                    print(f"item not expected: {item}")
                    verify_result = False
                
        return verify_result
        
def run_tests(tests: list[Test], func: Callable[[str], List[List[str]]]):
    for test in tests:
        result = func(test.s)
        if test.verify(result):
            print(f"Test passed for {test.s}")
        else:
            print(f"Test failed for {test.s}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test("s", [["s"]]),
    Test("mm", [["m","m"], ["mm"]]),
    Test("mom", [["m","o","m"], ["mom"]]),
    Test("noon", [["n","o","o","n"], ["n","oo","n"], ["noon"]]),
    Test("civic", [["c","i","v","i","c"], ["c","ivi","c"], ["civic"]]),
    Test("mommom", [["m","o","m","m","o","m"], ["m","ommo","m"], ["mom", "mom"], ["mommom"], ['m', 'o', 'm', 'mom'], ['m', 'o', 'mm', 'o', 'm'], ['mom', 'm', 'o', 'm'] ])
]

run_tests(tests, Solution().partition)
```
