### DESCRIPTION (inspired by Leetcode.com)

Given a reference head of type ListNode that is the head node of a singly linked list and an integer n, write a function that removes the n-th node from the end of the list and returns the head of the modified list.

Note: n is guaranteed to be between 1 and the length of the list. If n is the length of the list, the head of the list should be removed.

Example 1:

Input: n = 2

head -> 5 -> 4 -> 3 -> 2 -> 1

Output:

head -> 5 -> 4 -> 3 -> 1

Explanation: The 2nd to last node is removed from the list.

Example 2:

Input: n = 5

head -> 5 -> 4 -> 3 -> 2 -> 1

Output:

head -> 4 -> 3 -> 2 -> 1

Explanation: The 5th to last node is the head node, so it is removed.

```python
class ListNode:
    def __init__(self, val: int = 0, next: 'ListNode' = None):
        self.val = val
        self.next = next

    # convert node list to [], for easy display
    def toList(self) -> list[int]:
        current = self
        result = []
        while current:
            result.append(current.val)
            current = current.next

        return result

class Solution:
    def removeNthFromEnd(self, head: ListNode, n: int) -> ListNode:
        # 1. count the total number of nodes 
        current = head
        total = 0
        while current:
            total += 1
            current = current.next

        # 2. remove the (total-n)th node
        dummy = ListNode(0, head)
        prev = dummy
        current = prev.next
        for i in range(total-n):
            prev = current
            current = current.next

        prev.next = current.next
        return dummy.next
```

### Feedback

Your solution is correct and passed all tests. Counting the list first, then using a dummy node to remove the target handles removing the head cleanly, including when n == total. The loop runs total - n times, leaving prev immediately before the node to delete. 

Complexity is O(L) time and O(1) extra space. One interview-friendly improvement would be to avoid the unused current = prev.next initialization distinction by making the traversal intent explicit, and mention that current is guaranteed non-None because the problem guarantees valid n. 

```python
from typing import Callable

class Test:  
    def __init__(self, nodes: list[int], n: int, result: list[int]):
        self.input = nodes
        self.n = n
        self.expected_result = result
        
        current = ListNode()
        self.head = current
        
        for i in range(len(nodes)):
            current.next = ListNode(nodes[i])
            current = current.next

        self.head = self.head.next


def run_tests(tests: list[Test], func: Callable[[ListNode, int], bool]):
    for test in tests:
        test_output: ListNode = func(test.head, test.n)
        result = None
        if test_output:
            result = test_output.toList()
        
        if result == test.expected_result:
            print(f"Test passed for {test.input} with n: {test.n}")
        else:
            print(f"Test failed for {test.input} with n: {test.n}. Expected: {test.expected_result}, Actual: None")
```

```python
tests = [
    Test([5,4,3,2,1], 1, [5,4,3,2]),
    Test([5,4,3,2,1], 2, [5,4,3,1]),
    Test([5,4,3,2,1], 4, [5,3,2,1]),
    Test([5,4,3,2,1], 5, [4,3,2,1]),
    Test([5], 1, None),
    Test([5,2], 1, [5]),
    Test([5,2], 2, [2]),
]

run_tests(tests, Solution().removeNthFromEnd)
```
