### DESCRIPTION (inspired by Leetcode.com)
Write a function that takes an array of unsorted integers nums and an integer k, and returns the kth largest element in the array. This function should run in O(n log k) time, where n is the length of the array.

Example 1:

Inputs:

nums = [5, 3, 2, 1, 4]
k = 2

Output:

4

```python
import heapq

class Solution:
    def kthLargest(self, nums: list[int], k: int) -> int:
        # 1. build heap with the first k elements from list (k-known largets)
        # 2. for the remaining of the list (n-k), compare each with the root of heap.
        # If the new element is larger, pop the root node and push the new element to the heap.
        # 3. When all elements are checked, return the root node which is the Kth largest of the list.
        heap = nums[:k]
        heapq.heapify(heap)
        for num in nums[k:]:
            if num > heap[0]:
                heapq.heappop(heap)
                heapq.heappush(heap, num)

        return heap[0]
         
        
```

### Feedback

Your min-heap solution is correct and efficiently maintains the k largest values. It runs in O(n log k) time and uses O(k) extra space; duplicates and negative values are handled correctly. Since you initialize from nums[:k], this assumes valid input with 1 <= k <= len(nums), which is typical for the problem.

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Input:
    nums: list[int]
    k: int
    
@dataclass(frozen=True)
class Test:
    input: Input
    expected_result: int

def run_tests(tests: list[Test], func: Callable[[list[int], int], int]):
    for test in tests:
        result = func(test.input.nums, test.input.k)
        if result == test.expected_result:
            print("Test passed for " + str(test.input.nums))
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
```
