## Task

Design a parking lot system where different types of vehicles can park, and the system manages spot assignment and calculates fees upon exit.

#### Clarifying Questions
Q: what types of vehicles does the system need to support?
A: The system needs to support three vehicle types: Motorcycle, Car, and Large Vehicle.

Q: How does each type of vehicle fit in a parking spot? For example, can 2 motorcycles park on the same spot? Or does a large vehicle need 2 spots?
A: Each vehicle type fits in exactly one spot, and each spot holds exactly one vehicle at a time. So no sharing of spots between motorcycles, and a large vehicle occupies a single (larger) spot.

Q: So that means parking spot comes in 3 sizes, small, medium, and large?
A: Yes, that's correct. There are three sizes of parking spots — small for motorcycles, medium for cars, and large for large vehicles — and each spot size is only compatible with its corresponding vehicle type.

Q: If no small spots are available, can a larger spot be assigned to a small vehicle?
A: No — each spot size is only compatible with its corresponding vehicle type. A larger spot cannot be assigned to a smaller vehicle, and if no compatible spot is available, the system should reject the entry.

Q: What are the number of spots for each type?
A: The number of spots for each type is configurable at the time the parking lot is initialized — so your system should be able to accept any number of small, medium, and large spots as input.

Q: Is there any logic used for spot assignments?
A: The system should assign any available compatible spot — there's no specific ordering or priority logic required beyond finding an open spot that matches the vehicle type.

Q: How is the fee calculated?
A: The fee is calculated based on an hourly rate, rounded up to the nearest hour, and the same rate applies to all vehicle types regardless of spot size.

Q: so if a car has parked for 61 minutes, it will be charged for 2 hours?
A: Yes, that's correct. A vehicle parked for 61 minutes would be charged for 2 hours, since the fee is rounded up to the nearest hour.

Q: is there a maximum time a car can park?
A: There is no maximum parking duration — a vehicle can remain parked indefinitely, and the fee will continue to accrue for however long it stays.

Q: Does the system need to keep track of the total fee collected?
A: No, the system does not need to track total fees collected. Fee calculation only happens at the point of exit for an individual vehicle, and there's no reporting or aggregation requirement.

Q: Are all the spots the same? Are there some reserved for special needs?
A: All spots of the same size are equivalent — there are no reserved spots for special needs, disabilities, or any other purpose. Every spot of a given size is treated the same.

Q: Does the system need to implement payment part?
A: No, payment processing is out of scope for this system. You only need to calculate and return the fee amount at the time of exit — what happens after that (collecting payment, processing transactions, etc.) is not part of this design.

Q: Is a parking ticket necessary? Even without a parking ticket, the parking lot can still keep track of the vehicles' license number, the time of its entry and the spot assigned. The driver can check parking information related to his vehicle on a parking app without a ticket which can be lost or damaged.
A: For this design, a ticket is required. The system issues a ticket at entry and validates that ticket at exit — including rejecting exit if the ticket is invalid or has already been used. How you choose to model or implement that ticket internally is up to you.

Q: How will the system respond if the driver lost the parking ticket or it is damaged?
A: Lost ticket handling is explicitly out of scope for this system — you only need to handle the case where a ticket is invalid or has already been used.

Q: what does invalid mean? How can a ticket been used before?
Great questions. A ticket is considered invalid if it was never issued by the system — for example, if the ticket ID doesn't exist or was fabricated. A ticket is considered already used if the vehicle it was issued for has already exited — meaning the same ticket cannot be used a second time to exit again.

Q: Does the system need to distinguish between an invalid ticket and an used ticket? If yes, I will have to keep a copy of all used tickets in the design.
A: The system does not need to distinguish between the two in terms of the error message returned — both cases should simply result in a rejected exit. How you choose to track tickets internally to detect these conditions is up to you.

# 3. Class Design (1st attemmpt)
ParkingManager keeps a lists of parking spots for different vehicle types and a list of active tickets (cars still parking in spot). The constructor takes in parkingSpotCountMapping which is a key-value pair for VehicleType and its corresponding number of parking spots. This way we can support other vehicle types in the future.

When a new vehicle arrives, the entry() is called. It looks for available spot and if none is found, entry() will return false. Otherwise, it updates the state of the parking spot found, issues a ticket for it, adds the ticket to the active list and return true.

When a vehicle exits, the exit() is called. It validates the ticket by checking against the active list. If ticket is not found (validation fails), exit will return false. Otherwise, it calculates the parking fee based on the entry time in the ticket. It then removes the ticket from the active list and updates the status of the parking spot and returns true.  

ParkingSpot class is a data holder and has getters and setters.

Ticket class is also a data holder and has only getters. It stores the current datetime as vehicle entry time during initialization.

### Feedback
What went well

You've got a clean separation of responsibilities — ParkingManager owns the orchestration, while ParkingSpot and Ticket are data holders with getters. That's exactly the right instinct. 

Your entry and exit flows are well thought out. You explained the full lifecycle clearly: finding a spot, issuing a ticket, validating on exit, freeing the spot. That shows strong systems thinking. 

Keeping findAvailableSpot() as a private/internal method and only exposing entry() and exit() publicly is a great design choice — it keeps the interface clean. 

What needs improvement

Your entry() method currently returns a bool, but the caller also needs the ticket so the vehicle can present it at exit. Consider having it return a Ticket (or null/Optional on failure) instead — that way the ticket naturally flows out of the entry operation. 

Your Ticket class is missing vehicleType as a field. Right now you can look up the spot from the ticket, but if you ever want to extend pricing by vehicle type, or just have a complete audit record of the session, having it on the ticket is valuable.

Minor suggestions

You might consider storing activeTickets as a Map<ticketId, Ticket> rather than a Set<Ticket> — it makes the O(1) lookup during exit() validation more explicit and efficient.

# Class Design (2nd attempt)

ParkingManager keeps a lists of parking spots for different vehicle types and a list of active tickets (cars still parking in spot). The class also maintains separate queues for available spots for each vehicle type. This will make finding available spot in O(1) time.

When a new vehicle arrives, the enter() is called. It gets an available spot by poping from the available spots queue of the corresponding vehicle type. If queue is empty, enter() will return None. Otherwise, it issues a ticket for it, adds the ticket to the active list and return the ticket.

When a vehicle exits, the exit() is called. It validates the ticket by checking against the active list. If ticket is not found (validation fails), exit will return false. Otherwise, it calculates the parking fee based on the entry time in the ticket. It then removes the ticket from the active list, pushes the parking spot to the available queue and returns true.  

ParkingSpot and Ticket class are both data holders and have getters method.
