Imagine stepping stones across a river. You start on the first stone (index 0) and need to reach the final stone. Each stone has a number indicating the maximum distance you can leap from that position. Find the fewest number of leaps required to reach the end. You may assume a path to the end always exists.

Example 1

Input:
```
nums = [3, 4, 2, 1, 2, 1]
```

Output: 2

Explanation: From stone 0, leap to stone 1 (you could go up to 3 stones, but 1 is strategic). From stone 1, leap directly to stone 5 (4 stones forward). Total = 2 leaps.

Example 2

Input:
```
nums = [1, 2, 1, 1, 1]
```

Output: 3

Explanation: Leap 0→1, then 1→3, then 3→4. Three leaps total.

````python
# BRUTE FORCE, FAIL A TEST
from typing import List

class Solution:
    def jump(self, nums: List[int]) -> int:
        if not nums or len(nums) == 1:
            return 0
        
        min_leaps = float('inf')

        for i in range(len(nums)-1):
            j, leaps = i, i
            
            while j < len(nums)-1:
                if nums[j] == 0:
                    break
                j += nums[j]
                leaps += 1
            
            if j >= len(nums)-1:
                min_leaps = min(min_leaps, leaps)
                
        return min_leaps 
````

### Feedback
Your main issue is that you only follow jumps of exactly nums[j] and restart from every index. A jump can be any distance up to that value, so this misses shorter jumps that lead to a better route. For example, from index 0 in [5,9,3,...], jumping to index 1 is optimal; your code jumps directly to index 5. Also, starting leaps = i incorrectly assumes reaching index i costs i jumps, and the repeated simulation makes the complexity O(n²). Use a greedy level/range approach: track the farthest index reachable from the current jump range, and when you reach that range’s end, increment the jump count and start the next range. This runs in O(n) time and O(1) space.

````python
# WITH OPTIMIZATION, FAIL A TEST
from typing import List

class Solution:
    def jump(self, nums: List[int]) -> int:
        if not nums:
            return 0
        
        leaps = 0
        max_reach = 0

        for i in range(len(nums)-1):
            if i + nums[i] > max_reach:
                leaps += 1
                max_reach = i + nums[i]
                if max_reach >= len(nums)-1:
                    return leaps

        return leaps 
````

### Feedback

Your loop increments leaps whenever a single index extends max_reach, but jumps should be counted by layers/ranges, not by every improvement. For example, index 0 reaches index 7, so the first jump covers all indices 1–7; index 1 extending reach should not immediately count another jump. Track the end of the current jump range and the farthest reach available within it:

```
jumps = 0
current_end = 0
farthest = 0
for i in range(len(nums) - 1):
    farthest = max(farthest, i + nums[i])
    if i == current_end:
        jumps += 1
        current_end = farthest
return jumps
```
Your early return is fine, and the empty-input handling is harmless, but the key missing state is current_end. This greedy approach runs in O(n) time and O(1) space.

````python
# WITH OPTIMIZATION
from typing import List

class Solution:
    def jump(self, nums: List[int]) -> int:
        if not nums:
            return 0
        
        leaps = 0
        max_reach = 0
        current_end = 0

        for i in range(len(nums)-1):
            max_reach = max(max_reach, i + nums[i])

            if i == current_end:
                leaps += 1
                current_end = max_reach

        return leaps 
````

````python
from typing import Callable

class Test:  
    def __init__(self, input: List[int], result: bool):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[int]], bool]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
````
