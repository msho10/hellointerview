Given an array of distinct integers candidates and a target integer target, generate all unique combinations of candidates which sum to target. The combinations may be returned in any order, and the same number may be chosen from candidates an unlimited number of times.

Constraints:

All values in candidates are positive integers.
1 <= candidates.length <= 30
2 <= candidates[i] <= 40
All elements of candidates are distinct.
1 <= target <= 40

Input:
candidates = [2,3,6,7]
target = 7

Output: [[2,2,3],[7]]

Explanation:
2 and 3 are candidates, and 2 + 2 + 3 = 7. Note that 2 can be used multiple times. 7 is a candidate, and 7 = 7. These are the only two combinations.

```python
from typing import List

class Solution:
    def combinationSum(self, candidates: List[int], target: int) -> List[List[int]]:
        result = []
        nums = [n for n in candidates if n <= target]
        
        def dfs(path: List[int], remainder: int):
            if remainder == 0:
                sorted_path = sorted(path[:])
                if sorted_path not in result:
                    result.append(sorted_path)
            elif remainder > 0:
                for n in nums:
                    path.append(n)
                    dfs(path, remainder-n)
                    path.pop()
                            
            
        dfs([], target)
        return result
```

### Feedback

Your solution is correct and handles unlimited reuse, filtering, and the zero-remainder base case well. However, it explores every ordering of a combination: [2,3] and [3,2] are both generated, then deduplicated with sorted_path not in result. This causes substantial unnecessary work, with result membership checks and sorting adding further overhead. 

Maintain a start index and only choose numbers from that index onward; recurse with the same index to allow reuse. 

Sorting candidates first lets you stop when n > remainder, and each completed path is already sorted, so you can append it directly. 

Your current code is acceptable under these constraints, but the ordered backtracking approach is the key interview-level improvement.


```python
from typing import List

class Solution:
    def combinationSum(self, candidates: List[int], target: int) -> List[List[int]]:
        result = []
        candidates.sort()
        
        def dfs(path: List[int], remainder: int, start: int):
            if remainder == 0:
                result.append(path[:])
                return
            elif remainder > 0:
                for i in range(start, len(candidates)):
                    if candidates[i] > remainder:
                        return
                    path.append(candidates[i])
                    dfs(path, remainder-candidates[i], i)
                    path.pop()
                            
            
        dfs([], target, 0)
        return result
```

### Feedback

Your solution is correct and efficiently uses backtracking. Sorting enables the early return when candidates[i] > remainder, and passing i (rather than i + 1) correctly allows unlimited reuse. Starting each loop at start also prevents duplicate combinations, while combinations are generated in sorted order. Complexity is roughly exponential in the number of valid combinations, with recursion depth bounded by target / min(candidates). One interview improvement: explain why return is safe only because the candidates are sorted; without sorting, it would incorrectly skip later values. The elif remainder > 0 is harmless but unnecessary, since negative remainders cannot arise after the pruning check.

```python
from typing import Callable

class Input:
    def __init__(self, nums: List[int], target: int):
        self.nums = nums
        self.target = target
        
class Test:  
    def __init__(self, input: Input, result: List[List[int]]):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[int], int], List[List[int]]]):
    for test in tests:
        result = func(test.input.nums, test.input.target)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums}, {test.input.target}")
        else:
            print(f"Test failed for {test.input.nums}, {test.input.target}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input([2],1), []),
    Test(Input([2],2), [[2]]),
    Test(Input([2],3), []),  
    Test(Input([2,3,6,7], 7), [[2,2,3],[7]]),
    Test(Input([2,3,6,7,10,12], 10), [[2, 2, 2, 2, 2], [2, 2, 3, 3], [2, 2, 6], [3, 7], [10]]),
]

run_tests(tests, Solution().combinationSum)
```
