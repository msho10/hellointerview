You have to take a total of numCourses courses, which are labeled from 0 to numCourses - 1. You are given a list of prerequisites pairs, where prerequisites[i] = [a, b] indicates that you must complete course b before course a.

Given the total number of courses and a list of prerequisite pairs, write a function to return the ordering of courses you should take to finish all courses.

If there are multiple valid orderings, return any valid ordering. If it is impossible to finish all courses, return an empty array.

Example 1:

Input:

numCourses = 4
prerequisites = [[1,0], [2,0], [3,1], [3,2]]
Output: [0, 1, 2, 3] or [0, 2, 1, 3]

Explanation: There are a couple of ways to complete all courses, one possible order is [0, 1, 2, 3] and another is [0, 2, 1, 3].

Example 2:

Input:

numCourses = 2
prerequisites = [[1, 0], [0, 1]]
Output: []

Explanation: It is impossible to finish all courses, as you must finish course 0 before course 1 and course 1 before course 0.

```python
from typing import List
from collections import deque, defaultdict

class Solution:
    def findOrder(self, numCourses: int, prerequisites: List[List[int]]) -> List[int]:
        # 1. calculate indegree for each course and adjlist
        indegree = [0] * numCourses
        adjlist = defaultdict(list)
        for u, v in prerequisites:
            indegree[u] += 1
            adjlist[v].append(u)

        # 2. add course with 0 indegree to queue
        queue = deque([u for u in range(numCourses) if indegree[u] == 0])
        result = []

        # 3. while queue is not empty, pop each course, add to result list and decrement its neighbor. 
        # For neighbors that have 0 indegree, add to queue.        
        while queue:
            course = queue.popleft()
            result.append(course)
            for neighbor in adjlist[course]:
                indegree[neighbor] -= 1
                if indegree[neighbor] == 0:
                    queue.append(neighbor)

        return result if len(result) == numCourses else []
```

### Feedback
Your solution is correct and implements Kahn’s BFS topological sort cleanly. You build the edge direction correctly (v -> u), track indegrees, process all initially available courses, and detect cycles by comparing the result length with numCourses. It also handles edge cases such as no prerequisites and disconnected courses. 

Complexity is optimal: O(V + E) time and O(V + E) space. 

```python
from typing import Callable

class Input:
    def __init__(self, numCourses: int, prerequisites: List[List[int]]):
        self.prerequisites = prerequisites
        self.numCourses = numCourses
        
class Test:  
    def __init__(self, input: Input, result: List[int]):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[int, List[List[int]]], List[int]]):
    for test in tests:
        result = func(test.input.numCourses, test.input.prerequisites)
        if result == test.expected_result:
            print(f"Test passed for {test.input.prerequisites}")
        else:
            print(f"Test failed for {test.input.prerequisites}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input(1, []), [0]),
    Test(Input(2, [[1, 0]]), [0,1]),
    Test(Input(2, [[1, 0], [0, 1]]), []),
    Test(Input(3, [[1, 0], [2, 1]]), [0,1,2]),
    Test(Input(3, [[1, 0], [0, 1], [1, 2]]), []),
    Test(Input(4, [[1,0], [2,0], [3,1], [3,2]]), [0, 1, 2, 3]),
]

run_tests(tests, Solution().findOrder)
```
