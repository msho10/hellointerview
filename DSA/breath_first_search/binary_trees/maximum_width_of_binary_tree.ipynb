Given the root of a binary tree, write a function to find its maximum width. Each level of the binary tree has a width, which is the number of nodes between the leftmost and rightmost nodes at that level, including the null nodes between them. The function should return the maximum width of the binary tree.

Example 1:

Input: [4, 2, 7, 1, None, None, 9]

Output: 4

Explanation: The last level has 4 nodes [1, None, None, 9] which has the maximum width. 

Example 2:

Input: [4, 2, 7, 1]

Output: 2

Explanation: The third level only has one node, which means the width of that level is one. Level 2 has the maximum width, 2 nodes.

Example 3:

Input: [4, 2, 7, 1, None, 6, 9, 7, None, None, None, None, 1, 1, None]

Output: 7

Explanation: The last level has 7 nodes [7, None, None, None, None, 1, 1] which has the maximum width. 


```python
from typing import Optional
from collections import deque

class TreeNode:
    def __init__(self, val: int, left: Optional['TreeNode'] = None, right: Optional['TreeNode'] = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def maxWidth(self, root: Optional[TreeNode]) -> int:
        if root is None:
            return 0
        queue = deque([(root,0)])
        max_width = 0

        while queue:
            level_len = len(queue)
            left_most = right_most = -1
            current_width = 0
            
            for _ in range(level_len):
                node, index = queue.popleft()
                if left_most == -1:
                    left_most = index
                else:
                    right_most = index

                if node.left:
                    queue.append((node.left, index*2))
                if node.right:
                    queue.append((node.right, index*2+1))

            if left_most > -1 and right_most > -1:
                current_width = right_most - left_most + 1
            elif left_most > -1:
                current_width = 1

            max_width = max(max_width, current_width)

        return max_width        
```

### Feedback

Your solution is correct and passes the tests. The BFS correctly preserves each node’s conceptual position, so gaps between nodes contribute to the width. Your index*2/index*2+1 convention is valid as long as it is used consistently. The right_most initialization and the single-node-level handling are also correct. Complexity is O(n) time and O(n) space. 

One interview-quality improvement is to simplify width calculation by recording the first index and using the last processed index directly after each level; this removes the conditional current_width logic. 

For very deep, highly skewed trees, indices can become very large in Python, though normal constraints make this harmless; per-level index normalization can avoid that.


```python
from typing import Optional
from collections import deque

class TreeNode:
    def __init__(self, val: int, left: Optional['TreeNode'] = None, right: Optional['TreeNode'] = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def maxWidth(self, root: Optional[TreeNode]) -> int:
        if root is None:
            return 0
        queue = deque([(root,0)])
        max_width = 0

        while queue:
            level_len = len(queue)
            left_most = right_most = queue[0][1]
            
            for _ in range(level_len):
                node, index = queue.popleft()
                right_most = index

                if node.left:
                    queue.append((node.left, index*2))
                if node.right:
                    queue.append((node.right, index*2+1))

            max_width = max(max_width, right_most - left_most + 1)

        return max_width        
```

### Feedback

Correct solution—congratulations! Your BFS tracks conceptual heap indices, so missing nodes contribute to the width correctly, and processing exactly level_len nodes cleanly separates levels. Complexity is O(n) time and O(n) space. 

One interview-ready consideration: index can grow very large for a highly sparse, deep tree, since you never normalize indices per level. Python handles large integers, so correctness is preserved, but arithmetic and memory can grow unnecessarily. 

You can normalize each level by subtracting the first index from every position before generating children. Also, right_most is safely assigned because each processed level contains at least one node.



```python
from typing import Callable

class Input:
    def __init__(self, nums: list[int]):
        self.nums = nums
        
class Test:  
    def __init__(self, input: Input, result: int):
        self.input = input
        self.expected_result = result
        self.root = self._createNodefromList(input.nums, 0)

    def _createNodefromList(self, nums: list[int], index: int) -> TreeNode:
        if index >= len(nums) or nums[index] is None:
            return None
        newNode = TreeNode(nums[index])
        newNode.left = self._createNodefromList(nums, 2*index+1)
        newNode.right = self._createNodefromList(nums, 2*index+2)
        return newNode

def run_tests(tests: list[Test], func: Callable[[TreeNode], int]):
    for test in tests:
        result = func(test.root)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums}")
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input([]), 0),
    Test(Input([1]), 1),
    Test(Input([1,2]), 1),
    Test(Input([4, 2, 7, 1]), 2),
    Test(Input([4, 2, 7, 1, None, None, 9]), 4),
    Test(Input([4, 2, 7, 1, None, 6, 9, 7, None, None, None, None, 1, 1, None]), 7)
]

run_tests(tests, Solution().maxWidth)
```
