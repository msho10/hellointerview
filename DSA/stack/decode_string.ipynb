### DESCRIPTION (inspired by Leetcode.com)

Given an encoded string s, write a function to return its decoded string.

The encoding rule is k[encoded_string], where the encoded_string inside the square brackets is repeated exactly k times. k is always a positive integer, and the brackets can be nested.

You can assume the input is always well-formed: there are no extra spaces, every square bracket is properly matched, and digits only ever appear to specify a repeat count k (so you won't see input like 3a or 2[4]).

Constraints:

1 <= s.length
s consists of lowercase English letters, digits, and the square brackets [ and ].
All repeat counts k are positive integers and may have more than one digit (for example, 10[a]).
The input string is guaranteed to be valid.

Example 1:

Inputs:
s = "3[a]2[bc]"

Output:
"aaabcbc"

(Explanation: 3[a] decodes to "aaa" and 2[bc] decodes to "bcbc".)

Example 2:

Inputs:
s = "3[a2[c]]"

Output:
"accaccacc"

(Explanation: the inner 2[c] becomes "cc", so a2[c] is "acc", which is then repeated 3 times.)

Example 3:

Inputs:
s = "2[abc]3[cd]ef"

Output:
"abcabccdcdcdef"

```python
class Solution_V1:
    def decodeString(self, s: str) -> str:
  
        stack = []
        i = 0
        while i < len(s):
            while i < len(s) and s[i] != ']':
                stack.append(s[i])
                i += 1

            enc_str = ""
            while stack and stack[-1] != '[':
                enc_str = stack.pop() + enc_str
            
            if stack and stack[-1] == '[':
                stack.pop()
                k_str = ""
                while stack and str.isdigit(stack[-1]):
                    k_str = stack.pop() + k_str
                stack.append(int(k_str) * enc_str)
            else:
                stack.append(enc_str)

            i += 1

        dec_str = ""
        while stack:
            dec_str = stack.pop() + dec_str

        return dec_str 
            
```

### Feedback

Your solution is correct and passed the tests. It properly handles nested encodings, multi-digit repeat counts, and plain text by collapsing each completed ] segment on the stack. 

Complexity is generally O(n + output size), though repeated string prepending (enc_str = stack.pop() + enc_str and the final reconstruction) can cause extra copying and become less efficient for large inputs. 

In an interview, mention this tradeoff or build character lists and join them. 

Also, str.isdigit(stack[-1]) works, but stack[-1].isdigit() is more idiomatic and readable. 

The else branch after searching for [ is unnecessary for valid encoded input, since every ] should have a matching [. Overall, strong stack-based solution.

```python
class Solution:
    def decodeString(self, s: str) -> str:
        stack = []
        curr_str = ""
        k = 0

        for c in s:
            if c == '[':
                stack.append(curr_str)
                stack.append(k)
                curr_str = ""
                k = 0
            elif c == ']':
                repeat = stack.pop()
                prev_str = stack.pop()
                curr_str = prev_str + repeat * curr_str
            elif c.isdigit():
                k = k*10 + int(c)
            else:
                curr_str += c
                
        return curr_str
```

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Test:
    input: str
    expected_result: str

def run_tests(tests: list[Test], func: Callable[[str], str]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print("Test passed for " + str(test.input))
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test("",""),
    Test("a","a"),
    Test("ab","ab"),
    Test("1[a]", "a"),
    Test("2[a]", "aa"),
    Test("2[a]b", "aab"),
    Test("b2[a]", "baa"),
    Test("ab2[c]", "abcc"),
    Test("ab10[c]","abcccccccccc"),
    Test("3[a]2[bc]", "aaabcbc"),
    Test("3[a2[c]]", "accaccacc"),
    Test("2[abc]3[cd]ef", "abcabccdcdcdef")
]

run_tests(tests, Solution().decodeString)
```
