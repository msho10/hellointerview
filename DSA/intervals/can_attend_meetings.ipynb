# DESCRIPTION (inspired by Leetcode.com)
Write a function to check if a person can attend all the meetings scheduled without any time conflicts. Given an array intervals, where each element [s1, e1] represents a meeting starting at time s1 and ending at time e1, determine if there are any overlapping meetings. If there is no overlap between any meetings, return true; otherwise, return false.

Note that meetings ending and starting at the same time, such as (0,5) and (5,10), do not conflict.

Input:

intervals = [(1,5),(3,9),(6,8)]

Output:

false

Explanation: The meetings (1,5) and (3,9) overlap.

Input:

intervals = [(10,12),(6,9),(13,15)]

Output:

true

Explanation: There are no overlapping meetings, so the person can attend all.

```python
class Solution:
    def canAttendMeetings(self, intervals: list[list[int]]) -> bool:
        intervals.sort(key = lambda x:x[0])
        for i in range(1, len(intervals)):
            if intervals[i][0] < intervals[i-1][1]:
                return False
            
        return True
```

# Feedback

Correct solution — you sorted by start time and checked each meeting against the previous one, which is the right approach.

A couple of interview-quality notes:

- Your logic assumes touching intervals are allowed ([1,2] and [2,3]), which matches the strict < check. Good to call that out if asked.

- One subtle edge case: intervals.sort(...) mutates the input. That’s fine if allowed, but in interviews it’s worth mentioning or avoiding if you want to preserve the original list.

- Time complexity is O(n log n), space O(1) extra (ignoring sort internals).
Overall, this is clean and correct.

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Test:
    input: list[list[int]]
    expected_result: bool

def run_tests(tests: list[Test], func: Callable[[list[list[int]]], bool]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print("Test passed for " + str(test.input))
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test([(1,5),(3,9),(6,8)],False),
    Test([(10,12),(6,9),(13,15)],True)
]

run_tests(tests, Solution().canAttendMeetings)
```
