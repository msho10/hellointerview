Write a function to determine whether you can travel from the first index to the last index of an integer array nums, where each number in the array specifies the maximum distance you can jump from that index. The function should return true if reaching the last index is possible, and false otherwise.

Input:
```
nums = [1, 3, 0, 1, 4]
```
Output: true

Explanation: You can jump from index 0 to index 1, then jump from index 1 to index 4 which is the last index.

Example: 

Input:
```
nums = [2,2,1,0,5,1,1]
```
Output: false

Explanation: The first three indexes take you to index 3 no matter what. But you cannot move beyond index 3 because its maximum jump length is 0, making the indexes after index 3 unreachable.

````python
from typing import List

class Solution:
    def canJump(self, nums: List[int]) -> bool:
        
        i = len(nums)-2
        steps_needed = 1 # every move requires at least 1 step

        while i>-1:
            if nums[i] < steps_needed:
                steps_needed += 1
            else:
                steps_needed = 1

            i -= 1

        return steps_needed == 1
````

### Feedback

Your solution is correct and efficient: it uses a reverse greedy strategy, tracking how many consecutive steps are needed to reach the next known-good index. It runs in O(n) time and O(1) space. It also handles edge cases such as an empty array (returns True) and a single-element array (returns True), consistent with the usual definition that you start at the last index. The logic is compact, though steps_needed may be easier to understand if you describe it as the distance to the nearest reachable position. 

In an interview, briefly explain the invariant: scanning right to left, reset the required distance whenever the current index can reach the target; otherwise increase it.


````python
from typing import Callable

class Test:  
    def __init__(self, input: List[int], result: bool):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[int]], bool]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
````

````python
tests = [
    Test([1], True),
    Test([0], True),
    Test([1,1], True),
    Test([0,1], False),
    Test([1,0], True),
    Test([2,2,1,0,5,1,1], False),
    Test([1, 3, 0, 1, 4], True),
    Test([1, 3, 0, 1, 0, 2], False),
    Test([1, 3, 0, 1, 4, 0], True),
    Test([1, 1, 0, 0, 4], False),
    Test([1, 3, 0, 0, 4, 0], True),
    Test([5,9,3,2,1,0,2,3,3,1,0,0], True),
    Test([4, 0, 0, 0, 0], True),
    Test([4, 4, 0, 0, 0], True),
]

run_tests(tests, Solution().canJump)
````
