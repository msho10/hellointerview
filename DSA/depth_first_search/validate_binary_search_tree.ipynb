### DESCRIPTION (inspired by Leetcode.com)
Given the root of a binary tree, determine if it is a valid binary search tree (BST).

A tree is a BST if the following conditions are met:

Every node on the left subtree has a value less than the value of the current node.
Every node on the right subtree has a value greater than the value of the current node.
The left and right subtrees must also be valid BSTs.
Example 1:

Input:

[2,1,4]

Output: true

Example 2:

Input:

[4,1,5,null,null,3,6]

Output: false. 3 is the the root node's right subtree, but it is less than the root node 4.

```python
class TreeNode:
    def __init__(self, val: int, left: 'TreeNode' = None, right: 'TreeNode' = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def validateBST(self, root: TreeNode) -> bool:
        return self._dfs(root, -float('inf'), float('inf'))

    def _dfs(self, root: TreeNode, min_value: int, max_value: int) -> bool:
        if root is None:
            return True
        
        if root.val > min_value and root.val < max_value:
            return (self._dfs(root.left, min_value, root.val) and 
                    self._dfs(root.right, root.val, max_value))

        return False
            
                
```

### Feedback

Correct solution—congratulations. Your recursive bounds correctly enforce strict inequalities, so duplicates are rejected and constraints propagate through the entire subtree. Empty trees also return True. 

Complexity is O(n) time and O(h) recursion space, where h is tree height. 

One Python-specific edge case: using ±float('inf') can incorrectly reject a valid tree whose node value is positive or negative infinity, if such values are allowed. For typical integer inputs this is completely safe; otherwise, consider bounds that start as None and compare only when a bound exists. 

Also, annotate _dfs bounds as float/int or a suitable union rather than only int, since you pass float sentinels.

```python
from typing import Callable

class Input:
    def __init__(self, nums: list[int]):
        self.nums = nums
        
class Test:  
    def __init__(self, input: Input, result: bool):
        self.input = input
        self.expected_result = result
        self.root = self._createNodefromList(input.nums, 0)

    def _createNodefromList(self, nums: list[int], index: int) -> TreeNode:
        if index >= len(nums) or nums[index] is None:
            return None
        newNode = TreeNode(nums[index])
        newNode.left = self._createNodefromList(nums, 2*index+1)
        newNode.right = self._createNodefromList(nums, 2*index+2)
        return newNode

def run_tests(tests: list[Test], func: Callable[[TreeNode], bool]):
    for test in tests:
        result = func(test.root)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums}")
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input([]), True),
    Test(Input([1]), True),
    Test(Input([2,1]), True),
    Test(Input([2,None,5]), True),
    Test(Input([2,1,4]), True),
    Test(Input([4,1,5,None,None,3,6]), False)
]

run_tests(tests, Solution().validateBST)
```
