### Implementation for Adding Stop Request
The addStop() will call addHallCall() or addCabinRequest() depending on the request type. It should check if the floor requested is valid (exceeding range or cabin request is same as current floor) and if it is a hall call, make sure no up for top floor or down for ground floor. Once the request is confirmed valid, it will be added to one of the floor stops lists. 
If it is a hall call, the direction is provided. For up request, it will be added to _up_floor_stops, and for down request, it will be added to _down_floor_stops.
If it is a cabin request, the code will check if the request floor is less or greater than the current floor. If it is less, the elevator will have to move down in order to serve the request, so the request will go to _down_floor_stops. And vice verssa if the value is greater. 

In every Elevator.step(), if there is request in the current floor for the same direction as the elevator, it will be served. If the direction is different, it will be served after the elevator has switched its direction. This happens in the _move() which checks if the elevator is at the top floor or ground floor or there are no more requests ahead, but there are requests in the opposite direction, then a direction switch will take place. Once an elevator has switched its direction, the request that didn't get picked up in the last step() will be served in the next step().  

```python
## 4. Implementation
from enum import Enum
TOP_FLOOR = 9

class Direction(Enum):
    DOWN = 0
    UP = 1

class Elevator:
    def __init__(self, id: str):
        self._id = id
        self._currentFloor = 0
        self._direction = Direction.UP
        self._up_floor_stops = [0 for _ in range(TOP_FLOOR+1)]
        self._down_floor_stops = [0 for _ in range(TOP_FLOOR+1)]

    def addCabinRequest(self, floor: int):
        if not self._isValidFloor(floor):
            raise ValueError

        if floor == self._currentFloor:
            print("Already at floor")
            return
        
        if floor > self._currentFloor:
            self._up_floor_stops[floor] += 1
        else:
            self._down_floor_stops[floor] += 1
        print(f"Button pressed for floor {floor}")
        
    def addHallCall(self, floor: int, direction: Direction):
        if not self._isValidFloor(floor) or not self._isValidDirection(floor, direction):
            raise ValueError
        
        if direction == Direction.UP:
            self._up_floor_stops[floor] += 1
        else:
            self._down_floor_stops[floor] += 1

    def getDirection(self) -> Direction:
        return self._direction

    def getCurrentFloor(self) -> int:
        return self._currentFloor

    def getId(self) -> str:
        return self._id
    
    def getLastStop(self) -> int:
        if self._direction == Direction.UP:
            for i in range(TOP_FLOOR, self._currentFloor, -1):
                if self._up_floor_stops[i] > 0:
                    return i
        else:
            for i in range(0,self._currentFloor-1):
                if self._down_floor_stops[i] > 1:
                    return i
        return self._currentFloor

    def getStopsCountInBetween(self, from_floor: int, to_floor: int) -> int:
        if from_floor < 0 or from_floor > TOP_FLOOR or to_floor < 0 or to_floor > TOP_FLOOR:
            raise ValueError
        if from_floor == to_floor: return 0

        if from_floor > to_floor: # going down
            return sum(self._down_floor_stops[to_floor:from_floor+1])
        else:
            return sum(self._up_floor_stops[from_floor:to_floor+1])
        
     
    def _hasRequestsAhead(self) -> bool:
        if self._direction == Direction.UP and self._currentFloor < TOP_FLOOR:
            if sum(self._up_floor_stops[self._currentFloor + 1 :]) + sum(self._down_floor_stops[self._currentFloor + 1 :]) > 0:
                return True
        elif self._direction == Direction.DOWN and self._currentFloor > 0:
            if sum(self._up_floor_stops[:self._currentFloor]) + sum(self._down_floor_stops[:self._currentFloor]) > 0:
                return True
        
        return False
        
    def isIdle(self) -> bool:
        return (sum(self._up_floor_stops) + sum(self._down_floor_stops)) == 0

    def _isValidDirection(self, floor:int, direction: Direction) -> bool:
        if ((floor == TOP_FLOOR and direction == Direction.UP) or
            (floor == 0 and direction == Direction.DOWN)):
            return False
        return True
    
    def _isValidFloor(self, floor:int) -> bool:
        if floor < 0 or floor > TOP_FLOOR:
            return False
        return True
    
    def _openDoor(self):
        print(f"Opening door at floor {self._currentFloor}...")
        floor_request = input("Enter floor number to go to: ")
        if floor_request.isnumeric():
            self.addCabinRequest(int(floor_request))

    def _move(self):
        # No requests, no moves
        if self.isIdle():
            print("Idle at floor: " + str(self._currentFloor))         
            return

        if self._direction == Direction.UP:
            # switch direction if already at top floor or there are no more requests ahead                   
            if self._currentFloor == TOP_FLOOR or not self._hasRequestsAhead():
                print(f"Switching direction to DOWN at floor {self._currentFloor}")
                self._direction = Direction.DOWN
            else:
                self._currentFloor += 1
                print("Ascending to floor " + str(self._currentFloor))
                        
            
        elif self._direction == Direction.DOWN:
            # switch direction if already at ground floor or there are no more requests ahead
            if self._currentFloor == 0 or not self._hasRequestsAhead():
                print(f"Switching direction to UP at floor {self._currentFloor}")
                self._direction = Direction.UP
            else:
                self._currentFloor -= 1
                print("Descending to floor " + str(self._currentFloor))

    def step(self):
        # Check if there are requests for the current floor matching current direction.
        # Don't open door if direction is not the same (for hall calls) 
        # If so, serve them and remove request from list.
        if self._direction == Direction.UP and self._up_floor_stops[self._currentFloor] > 0:
            self._openDoor()
            self._up_floor_stops[self._currentFloor] = 0

        if self._direction == Direction.DOWN and self._down_floor_stops[self._currentFloor] > 0:
            self._openDoor()
            self._down_floor_stops[self._currentFloor] = 0

        self._move()
         
    
```

### Implementation for Dispatch Logic
I am using SCAN logic (elevator algorithm). Dispatch implementation will calculate the cost for these 3 conditions and
assign the elevator with the lowest cost:
1. Elevator is idle
2. Hall call is in the same direction as the travelling elevator and the request floor is ahead of 
the elevator's current floor.
3. Hall calls is in the opposite direction than the elevator's current direction or the elevator has already passed the
request floor. The elevator will need to first complete the trip in the current direction, then switch direction and travel to the request floor. 

Also, I have added the methods getLastStop() and getStopsCountInBetween() in the Elevator class in order to provide the Controller class with enough information to calculate the cost for each elevator.

```python
from enum import IntEnum
class Penalty(IntEnum):
    FLOOR_STOP = 3
    SWITCH_DIRECTION = 5

class Controller:
    def __init__(self, elevators: list[Elevator]):
        self._elevators = elevators
       
    def dispatchElevator(self, floor: int, direction: Direction):
        if (floor < 0 or floor > TOP_FLOOR or
            (direction == Direction.UP and floor == TOP_FLOOR)
            or (direction == Direction.DOWN and floor == 0 )):
            print("Invalid hall call")
            return
        
        min_cost = float('inf')
        elevator_index = 0

        for i, elevator in enumerate(self._elevators):
            cost = self._calculateCost(floor, direction, elevator)
            if cost < min_cost:
                min_cost = cost
                elevator_index = i

        self._elevators[elevator_index].addHallCall(floor, direction)
        print(f"Elevator {elevator_index} dispatched for hall call at floor {floor} going {direction.name}")
                


    def _calculateCost(self, requestFloor: int, requestDirection: Direction, elevator: Elevator) -> int:
        # Condition 1: elevator is idle
        if elevator.isIdle():
            return abs(elevator.getCurrentFloor() - requestFloor)

        # Condition 2: elevator is in the same direction as hall call and request floor is ahead of elevator's current floor
        if elevator.getDirection() == requestDirection:
            if (requestDirection == Direction.UP and requestFloor >= elevator.getCurrentFloor() or
                requestDirection == Direction.DOWN and requestFloor <= elevator.getCurrentFloor()):
                penalty = Penalty.FLOOR_STOP * elevator.getStopsCountInBetween(elevator.getCurrentFloor(), requestFloor)
                return abs(requestFloor - elevator.getCurrentFloor()) + penalty         

        # Condition 3: elevator is in opposite direction as hall call or elevator has already passed the request floor
        last_stop = elevator.getLastStop()
        penalty = Penalty.FLOOR_STOP * elevator.getStopsCountInBetween(elevator.getCurrentFloor(), last_stop)
        penalty += Penalty.FLOOR_STOP * elevator.getStopsCountInBetween(last_stop, requestFloor)
        penalty += Penalty.SWITCH_DIRECTION
        stops_in_current_direction = abs(last_stop - elevator.getCurrentFloor())
        stops_in_opposite_direction = abs(last_stop - requestFloor)
        return stops_in_current_direction + stops_in_opposite_direction + penalty
    
    
    def step(self):
        for i, elev in enumerate(self._elevators):
            print("---------------------")
            print(f"ELEVATOR {i} ")
            elev.step()
            print("---------------------")                  
            print("")
        
        
```

## 4. Extensibility

Suppose you want to support different dispatching strategies — for example, swapping your current cost-based SCAN algorithm in _calculateCost() for a shortest-seek-time-first (SSTF) strategy that simply picks the elevator closest to the request floor, ignoring stops in between. How would you restructure your Controller class to make the scheduling algorithm swappable without modifying Controller itself each time?

I will use strategy pattern to achieve this. So instead of having the _calculateCost() implementation inside the Controller class, the class will have a reference to a DispatchStrategy abstract class and the actual implementation will be passed to it during runtime. This way I could have different implementations of the abstract class (e.g. SCAN or SSTF) and could swap between them without modifying the Controller itself.

One small thing to consider: you initialize _dispatchStrategy = None and throw an error if it's not set — a slightly cleaner pattern is to inject the strategy through the constructor (def __init__(self, elevators, strategy: DispatchStrategy)) so the Controller is never in an invalid state right after creation. That way you don't need the setDispatchStrategy() setter at all unless you explicitly want runtime swapping. 

```python
from enum import IntEnum
from abc import ABC, abstractmethod

class DispatchStrategy(ABC):
    @abstractmethod
    def findTheBestElevator(self, floor: int, direction: Direction, elevators: list[Elevator]) -> Elevator:
        pass

class Penalty(IntEnum):
    FLOOR_STOP = 3
    SWITCH_DIRECTION = 5

class SCAN(DispatchStrategy):
    def __init__(self):
        super().__init__()

    def findTheBestElevator(self, floor: int, direction: Direction, elevators: list[Elevator]) -> Elevator:
        min_cost = float('inf')
        best_elevator = elevators[0]

        for elevator in elevators:
            cost = self._calculateCost(floor, direction, elevator)
            if cost < min_cost:
                min_cost = cost
                best_elevator = elevator

        return best_elevator

    def _calculateCost(self, requestFloor: int, requestDirection: Direction, elevator: Elevator) -> int:
            # Condition 1: elevator is idle
            if elevator.isIdle():
                return abs(elevator.getCurrentFloor() - requestFloor)
    
            # Condition 2: elevator is in the same direction as hall call and request floor is ahead of elevator's current floor
            if elevator.getDirection() == requestDirection:
                if (requestDirection == Direction.UP and requestFloor >= elevator.getCurrentFloor() or
                    requestDirection == Direction.DOWN and requestFloor <= elevator.getCurrentFloor()):
                    penalty = Penalty.FLOOR_STOP * elevator.getStopsCountInBetween(elevator.getCurrentFloor(), requestFloor)
                    return abs(requestFloor - elevator.getCurrentFloor()) + penalty         
    
            # Condition 3: elevator is in opposite direction as hall call or elevator has already passed the request floor
            last_stop = elevator.getLastStop()
            penalty = Penalty.FLOOR_STOP * elevator.getStopsCountInBetween(elevator.getCurrentFloor(), last_stop)
            penalty += Penalty.FLOOR_STOP * elevator.getStopsCountInBetween(last_stop, requestFloor)
            penalty += Penalty.SWITCH_DIRECTION
            stops_in_current_direction = abs(last_stop - elevator.getCurrentFloor())
            stops_in_opposite_direction = abs(last_stop - requestFloor)
            return stops_in_current_direction + stops_in_opposite_direction + penalty
        
        
class Controller:
    def __init__(self, elevators: list[Elevator], strategy: DispatchStrategy):
        self._elevators = elevators
        self._dispatchStrategy = strategy

    def dispatchElevator(self, floor: int, direction: Direction):
        if not self._dispatchStrategy:
            raise ValueError("Dispatch strategy is not set.") 
        
        if (floor < 0 or floor > TOP_FLOOR or
            (direction == Direction.UP and floor == TOP_FLOOR)
            or (direction == Direction.DOWN and floor == 0 )):
            print("Invalid hall call")
            return
        
        elevator = self._dispatchStrategy.findTheBestElevator(floor, direction, self._elevators)
        elevator.addHallCall(floor, direction)
        print(f"Elevator {elevator.getId()} dispatched for hall call at floor {floor} going {direction.name}")
                
    
    def step(self):
        for elevator in self._elevators:
            print("---------------------")
            print(f"ELEVATOR {elevator.getId()} ")
            elevator.step()
            print("---------------------")                  
            print("")
        
        
```
