### DESCRIPTION (inspired by Leetcode.com)
Given the root of a binary tree and an integer target, write a recursive function to find all root-to-leaf paths where the sum of all the values along the path sum to target.

Example 1:

Input:
[1,2,4,4,7,5,1]
target = 10

Output:
[[1,2,7],[1,4,5]] # [[1,4,5],[1,2,7]] is also accepted.

The paths are 1 -> 2 -> 7 and 1 -> 4 -> 5

```python
class TreeNode:
    def __init__(self, val: int, left: 'TreeNode' = None, right: 'TreeNode' = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def pathSum(self, root: TreeNode, target: int) -> list[list[int]]:
        result = []

        def dfs(node: TreeNode, path: list[int], target: int):
            if node is None:
                return
            
            path.append(node.val)
            target -= node.val 
            if node.left is None and node.right is None:
                if target == 0:
                    result.append(path)
            else:
                dfs(node.left, path.copy(), target)
                dfs(node.right, path.copy(), target)
            
            
        dfs(root, [], target)
        return result
```

### Feedback
Your solution is correct and handles empty trees, negative values, and leaf-only matching properly. Using path.copy() ensures sibling branches do not interfere, and the time complexity is O(n·h) in the worst case due to copying paths (with O(n·h) result/copy space). 

A useful optimization is to backtrack with one shared list—append before recursion, then pop() afterward—and append path.copy() only when a match is found. This reduces copying during traversal and typically gives O(n) auxiliary space (excluding returned paths). Be careful not to append the mutable path itself to result, or later backtracking would alter stored answers.

```python
from typing import Callable

class Input:
    def __init__(self, nums: list[int], target: int):
        self.nums = nums
        self.target = target
        
class Test:  
    def __init__(self, input: Input, result: list[list[int]]):
        self.input = input
        self.expected_result = result
        self.root = self._createNodefromList(input.nums, 0)

    def _createNodefromList(self, nums: list[int], index: int) -> TreeNode:
        if index >= len(nums) or nums[index] is None:
            return None
        newNode = TreeNode(nums[index])
        newNode.left = self._createNodefromList(nums, 2*index+1)
        newNode.right = self._createNodefromList(nums, 2*index+2)
        return newNode

def run_tests(tests: list[Test], func: Callable[[TreeNode, int], list[list[int]]]):
    for test in tests:
        result = func(test.root, test.input.target)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums}")
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input([],1),[]),
    Test(Input([2],1),[]),
    Test(Input([3],3),[[3]]),
    Test(Input([1,3],4),[[1,3]]),
    Test(Input([1,None,3],4),[[1,3]]),
    Test(Input([-1],-1),[[-1]]),
    Test(Input([1,2,3],0),[]),
    Test(Input([1,2,4,4,7,5,1],10),[[1,2,7],[1,4,5]])    
]

run_tests(tests, Solution().pathSum)
```
