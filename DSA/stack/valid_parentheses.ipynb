### DESCRIPTION (inspired by Leetcode.com)

Given an input string s consisting solely of the characters '(', ')', '{', '}', '[' and ']', determine whether s is a valid string. A string is considered valid if every opening bracket is closed by a matching type of bracket and in the correct order, and every closing bracket has a corresponding opening bracket of the same type.

Example 1:

Inputs:
s = "(){({})}"

Output:
True

Example 2:

Inputs:
s = "(){({}})"

Output:
False

```python
class Solution_V1:
    def isValid(self, s: str) -> bool:
        if not s:
            return True

        parenthesis = {')':'(' , ']':'[' , '}':'{'}
        stack = []

        for c in s:
            if c == '(' or c == '{' or c == '[':
                stack.append(c)
            elif not stack:  # empty stack
                return False
            else:
                if stack.pop() != parenthesis[c]:
                    return False
            
        return True if not stack else False
```

### Feedback

Your solution is correct and passed all tests. The stack correctly enforces matching types and nesting order, and you handle empty input, premature closers, mismatches, and leftover openers. Complexity is optimal: O(n) time and O(n) space. One small readability improvement is to return not stack instead of return True if not stack else False. Also, because the problem guarantees only bracket characters, indexing parenthesis[c] is safe; without that guarantee, an unexpected character would raise KeyError. Your explicit opening-bracket checks are clear and valid for this constrained input.

```python
class Solution:
    def isValid(self, s: str) -> bool:
        if not s:
            return True

        parenthesis = {')':'(' , ']':'[' , '}':'{'}
        stack = []

        for c in s:
            if c == '(' or c == '{' or c == '[':
                stack.append(c)
            elif not stack:  # empty stack
                return False
            else:
                if c not in parenthesis:
                    raise ValueError
                if stack.pop() != parenthesis[c]:
                    return False
            
        return not stack
```

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Test:
    input: str
    expected_result: bool

def run_tests(tests: list[Test], func: Callable[[str], bool]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print("Test passed for " + str(test.input))
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test("", False),
    Test("(", False),
    Test("}", False),
    Test("(}", False),
    Test("[]", True),
    Test("((", False),
    Test("())", False),
    Test("()}", False),
    Test("({})", True),
    Test("()[]", True),
    Test("(){({})}", True),
    Test("(){({}})", False),
]

run_tests(tests, Solution().isValid)
```
