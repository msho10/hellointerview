You are given an m x n binary matrix grid where each cell contains either a 0 or a 1.

Write a function that returns a matrix of the same dimensions where each cell contains the distance to the nearest 0 in the original matrix. The distance between two adjacent cells is 1. If there is no 0 in the grid, return -1 for each cell.

Example 1:
```
mat = [
  [1, 0, 1],
  [0, 1, 0],
  [1, 1, 1],
]

Output:
[
  [1, 0, 1],
  [0, 1, 0],
  [1, 2, 1],
]
```

````python
from typing import List
from collections import deque

class Solution_V1:
    def updateMatrix(self, mat: List[List[int]]) -> List[List[int]]:
        if not mat or not mat[0]:
            return mat

        rows, cols = len(mat), len(mat[0])
        result = [[0 for _ in range(cols)] for _ in range(rows)]

        def getDistance(r, c) -> int:
            queue = deque([(r, c)])
            visited = {(r, c)}
            distance = 0
            directions = [(0,-1), (-1,0), (0,1), (1,0)]

            while queue:
                distance += 1
                queue_len = len(queue)
                for _ in range(queue_len):
                    r_, c_ = queue.popleft()                
                    for dr, dc in directions:
                        nr = dr + r_
                        nc = dc + c_
                        if 0 <= nr < rows and 0 <= nc < cols:
                            if mat[nr][nc] == 0:
                                return distance                                                
                            if (nr, nc) not in visited:                        
                                queue.append((nr, nc))
                                visited.add((nr, nc))

            return -1         
                                
        for r in range(rows):
            for c in range(cols):
                if mat[r][c] == 1:
                   result[r][c] = getDistance(r,c)
                        
        return result
````

### Feedback

Your solution is correct and handles empty input, unreachable cells, and the all-zero case. The per-cell BFS also correctly returns the nearest distance. 

However, its worst-case complexity is O((mn)^2): you may traverse most of the matrix separately for every 1, with additional per-search memory up to O(mn). 

For an interview, use multi-source BFS: enqueue every 0 once, then expand outward while recording distances. That reduces time to O(mn) and space to O(mn). Minor readability improvement: define directions once outside getDistance, since recreating it for every cell is unnecessary.

````python
from typing import List
from collections import deque

class Solution:
    def updateMatrix(self, mat: List[List[int]]) -> List[List[int]]:
        if not mat or not mat[0]:
            return mat

        rows, cols = len(mat), len(mat[0])
        result = [[-1] * cols for _ in range(rows)]
        directions = [(0,-1), (-1,0), (0,1), (1,0)]
        queue = deque()
        
        # find all 0's in the input matrix and add them to the queue
        for r in range(rows):
            for c in range(cols):
                if mat[r][c] == 0:
                    queue.append((r,c))
                    result[r][c] = 0

        # update the -1's in the result matrix with distance in each iteration
        distance = 1
        while queue:
            queue_len = len(queue)
            for _ in range(queue_len):
                r, c = queue.popleft()
                for dr, dc in directions:
                    nr, nc = r + dr, c + dc
                    if 0 <= nr < rows and 0 <= nc < cols:
                        if result[nr][nc] == -1: 
                            result[nr][nc] = distance
                            queue.append((nr,nc))
            distance += 1
                        
        return result
````

````python
from typing import Callable

class Input:
    def __init__(self, grid: List[List[str]]):
        self.grid = grid
        
class Test:  
    def __init__(self, input: Input, result: List[List[int]]):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[List[str]]], List[List[int]]]):
    for test in tests:
        result = func(test.input.grid)
        if result == test.expected_result:
            print(f"Test passed for {test.input.grid}")
        else:
            print(f"Test failed for {test.input.grid}. Expected: {test.expected_result}, Actual: {result}")
````
