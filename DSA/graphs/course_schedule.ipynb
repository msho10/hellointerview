You have to take a total of numCourses courses, which are labeled from 0 to numCourses - 1. You are given a list of prerequisites pairs, where prerequisites[i] = [a, b] indicates that you must complete course b before course a.

Given the total number of courses and a list of prerequisite pairs, write a function to determine if it is possible to finish all courses.

Example 1:

Input:
numCourses = 3

prerequisites = [[1, 0], [2, 1]]
Output: true

Explanation: You can take courses in the following order: 0, 1, 2.

Example 2:

Input:
numCourses = 3

prerequisites = [[1, 0], [0, 1], [1, 2]]
Output: false

Explanation: It is impossible to finish all courses, as you must finish course 1 before course 0 and course 0 before course 1.

```python
from typing import List
from collections import deque, defaultdict

class Solution:
    def canFinish(self, numCourses: int, prerequisites: List[List[int]]) -> bool:
        # 1. Calculate the indegree and adjlist for the prerequisites
        indegree = [0] * numCourses
        adjlist = defaultdict(list)
        for u, v in prerequisites:
            indegree[u] += 1
            adjlist[v].append(u)

        # 2. Put in queue for those with 0 indegree
        queue = deque([u for u in range(numCourses) if indegree[u] == 0])
        order = []

        # 3. Until the queue is empty, pop from the queue, add it to the topological order, decrement the indegree for its neighbors
        while queue:
            u = queue.popleft()
            order.append(u)
            for neighbor in adjlist[u]:
                indegree[neighbor] -= 1
                if indegree[neighbor] == 0:
                    queue.append(neighbor)

        return len(order) == numCourses
```

### Feedback

Correct solution—congratulations! You implemented Kahn’s topological-sort algorithm accurately: each prerequisite edge contributes to the dependent course’s indegree, and processing zero-indegree courses detects cycles. Returning len(order) == numCourses also correctly handles disconnected courses and the numCourses == 0 edge case. 

Complexity is optimal: O(numCourses + len(prerequisites)) time and space. 

One practical detail: ensure defaultdict and deque are imported (for example, from collections import defaultdict, deque) unless the interview environment provides them. 

order is valid for tracking processed courses; a counter could use slightly less memory, but this is not important here.

```python
from typing import Callable

class Input:
    def __init__(self, numCourses: int, prerequisites: List[List[int]]):
        self.prerequisites = prerequisites
        self.numCourses = numCourses
        
class Test:  
    def __init__(self, input: Input, result: bool):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[int, List[List[int]]], bool]):
    for test in tests:
        result = func(test.input.numCourses, test.input.prerequisites)
        if result == test.expected_result:
            print(f"Test passed for {test.input.prerequisites}")
        else:
            print(f"Test failed for {test.input.prerequisites}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input(1, []), True),
    Test(Input(2, [[1, 0]]), True),
    Test(Input(2, [[1, 0], [0, 1]]), False),
    Test(Input(3, [[1, 0], [2, 1]]), True),
    Test(Input(3, [[1, 0], [0, 1], [1, 2]]), False),
]

run_tests(tests, Solution().canFinish)
```
