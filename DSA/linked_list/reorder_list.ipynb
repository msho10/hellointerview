### DESCRIPTION (inspired by [Leetcode.com](https://leetcode.com/problems/reorder-list))

Given a reference head of type ListNode that is the head of a singly linked list, reorder the list **in-place** such that the nodes are reordered to form the following pattern:

1st node -> last node -> 2nd node -> 2nd to last node -> 3rd node ...

**Example 1:** 

input:

head -> 5 -> 4 -> 3 -> 2 -> 1

output:

head -> 5 -> 1 -> 4 -> 2 -> 3

**Example 2:** 

input:

 head -> 0 -> 1 -> 2

output:

 head -> 0 -> 2 -> 1

```python
class ListNode:
    def __init__(self, val: int = 0, next: 'ListNode' = None):
        self.val = val
        self.next = next
        
    # convert node list to [], for easy display
    def toList(self) -> list[int]:
        current = self
        result = []
        while current:
            result.append(current.val)
            current = current.next

        return result
    
class Solution:
    def reorderList(self, head: ListNode) -> None:
        # approach:
        # 1. use slow/fast pointer to get to the middle and end of the list 
        # 2. reverse the order of second half of the list, so that we can traverse 
        # the nodes easilyl during reordering.
        # 3. reorder the list by picking one node from the first half of the list and 
        # one node from the second half of the list, increment the pointers for both list
        # 4. continue until the 2 pointers converge

        if not head or not head.next:
            return
        
        # 1. slow/fast pointer to get to middle of list
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        # 2. reverse the second half of list
        prev = None
        current = slow
        while current:
            _next = current.next
            current.next = prev
            prev = current
            current = _next

        # 3. reorder by merging the 2 lists
        l1 = head
        l2 = prev
       
        while l2.next:
            next_l1 = l1.next
            next_l2 = l2.next
            l1.next = l2
            l2.next = next_l1
            l1 = next_l1
            l2 = next_l2
                        
```

```python
from typing import Callable
class Test:  
    def __init__(self, nodes: list[int], result: list[int]):
        self.input = nodes
        self.expected_result = result
        
        current = ListNode()
        self.head = current
        
        for i in range(len(nodes)):
            current.next = ListNode(nodes[i])
            current = current.next

        self.head = self.head.next

def run_tests(tests: list[Test], func: Callable[[ListNode], None]):
    for test in tests:
        func(test.head)
        actual_result = []
        if test.head:
            actual_result = test.head.toList()
            
        if  actual_result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {actual_result}")
```

```python
tests = [
    Test([],[]),
    Test([0], [0]),
    Test([0,1], [0,1]),
    Test([0,1,2], [0,2,1]),
    Test([0,1,2,3], [0,3,1,2]),
    Test([0,1,2,3,4], [0,4,1,3,2]),
]

run_tests(tests, Solution().reorderList)
```
