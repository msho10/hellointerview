### DESCRIPTION (inspired by Leetcode.com)
Given the root of the binary tree, find the longest path where all nodes along the path have the same value. This path doesn't have to include the root node, and it can bend at a node to connect that node's left and right subtrees (as in Example 2 below). Return the number of edges on that path, not the number of nodes.

Example 1:

Input:

[1,4,5,4,4,5]
Output: 2

Explanation: The longest path of the same value is the path [4,4,4], which has a total of 2 edges.

Example 2:

Input:

[1,1,1,1,1,1,1]
Output: 4

Explanation: The longest path of the same value is the path [1,1,1,1,1], which has a length of 4.

```python
class TreeNode:
    def __init__(self, val: int, left: 'TreeNode' = None, right: 'TreeNode' = None):
        self.val = val
        self.left = left
        self.right = right

class Solution_V1:
    def longestUnivaluePath(self, root: TreeNode) -> int:
        max_ = 0

        def dfs(node: TreeNode) -> int:
            if node is None:
                return 0
            
            nonlocal max_
            left = dfs(node.left)
            right = dfs(node.right)

            if node.left and node.val == node.left.val:
                left += 1
            if node.right and node.val == node.right.val:
                right += 1

            if node.left and node.right and (node.val == node.left.val == node.right.val):
                max_ = max(max_, left + right)
            else:
                max_ = max(max_, left, right)
                
            return max(left, right)

        dfs(root)
        return max_
```

Correct solution—congratulations. Your postorder DFS correctly computes the longest same-value downward path from each child, updates the global answer for paths that bend through the current node, and returns the best single branch to the parent. It also handles None and single-node trees correctly, with O(n) time and O(h) recursion space. 

One small readability improvement: compute left and right as zero unless the corresponding child matches, rather than adding to the raw DFS results and then conditionally using them. Your current logic is still correct because max(left, right) discards nonmatching branches, but explicit “arrow” variables make that invariant clearer and reduce the chance of mistakes in variations of this problem.

```python
class TreeNode:
    def __init__(self, val: int, left: 'TreeNode' = None, right: 'TreeNode' = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def longestUnivaluePath(self, root: TreeNode) -> int:
        max_ = 0

        def dfs(node: TreeNode) -> int:
            if node is None:
                return 0
            
            nonlocal max_
            left_subtree = dfs(node.left)
            right_subtree = dfs(node.right)
            left_path = right_path = 0

            if node.left and node.val == node.left.val:
                left_path = 1 + left_subtree
            if node.right and node.val == node.right.val:
                right_path = 1 + right_subtree

            max_ = max(max_, left_path + right_path)
            return max(left_path, right_path)

        dfs(root)
        return max_
```

```python
from typing import Callable

class Input:
    def __init__(self, nums: list[int]):
        self.nums = nums
        
class Test:  
    def __init__(self, input: Input, result: int):
        self.input = input
        self.expected_result = result
        self.root = self._createNodefromList(input.nums, 0)

    def _createNodefromList(self, nums: list[int], index: int) -> TreeNode:
        if index >= len(nums) or nums[index] is None:
            return None
        newNode = TreeNode(nums[index])
        newNode.left = self._createNodefromList(nums, 2*index+1)
        newNode.right = self._createNodefromList(nums, 2*index+2)
        return newNode

def run_tests(tests: list[Test], func: Callable[[TreeNode], int]):
    for test in tests:
        result = func(test.root)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums}")
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input([]),0),
    Test(Input([4]),0),
    Test(Input([1,2]),0),
    Test(Input([1,2,3]),0),
    Test(Input([1,2,2,4]),0),
    Test(Input([1,1,1,4]),2),
    Test(Input([1,1,2,4]),1),
    Test(Input([1,0,1,4]),1),
    Test(Input([1,4,5,4,4,5]),2),
    Test(Input([1,1,1,1,1,1,1]),4),
        
    
]

run_tests(tests, Solution().longestUnivaluePath)
```
