### DESCRIPTION (inspired by Leetcode.com)
You're given a square grid (n × n matrix) where each row is sorted in ascending order from left to right, and each column is also sorted in ascending order from top to bottom.

Given the matrix and an integer k, find the k-th smallest element among all elements in the matrix.

Note: k is 1-indexed, meaning k = 1 returns the smallest element.

Example 1:

Input:

matrix = [
    [ 1, 5, 9],
    [10,11,13],
    [12,13,15]]
k = 8
Output: 13

Explanation: The elements in sorted order are [1,5,9,10,11,12,13,13,15]. The 8th smallest element is 13.

Example 2:

Input:

matrix = [[-5]], k = 1
Output: -5

Explanation: The matrix has only one element, so it's the 1st smallest.

```python
class Solution:
    def kthSmallest(self, matrix: list[list[int]], k: int) -> int:
        array = []
        for r in matrix:
            array.extend(r)
        array.sort()

        if len(array) < k:
            raise ValueError(f"Matrix has {len(array)} elements which is less than k = {k}")
        
        return array[k-1]
```

### Feedback

Your solution is correct and handles duplicates and negative values properly. Flattening, sorting, and returning array[k - 1] gives the right result, and your explicit k validation is a useful defensive check. 

The main interview consideration is efficiency: for an n × n matrix, this uses O(n²) extra space and O(n² log n) time, ignoring the matrix’s sorted structure. You can improve this to O(1) extra space and O(n log(value range)) time with value-based binary search, counting values ≤ mid using the row/column ordering. 

Also consider validating k >= 1 if inputs are not guaranteed valid; otherwise k = 0 would incorrectly access the last element.

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Input:
    matrix: list[list[int]]
    k: int
    
@dataclass(frozen=True)
class Test:
    input: Input
    expected_result: int

def run_tests(tests: list[Test], func: Callable[[list[list[int]], int], int]):
    for test in tests:
        result = func(test.input.matrix, test.input.k)
        if result == test.expected_result:
            print("Test passed for " + str(test.input.matrix))
        else:
            print(f"Test failed for {test.input.matrix}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests =[ 
    Test(Input([[-5]],1),-5),
    Test(Input([[1,5],[4,6]],2),4),
    Test(Input([[ 1, 5, 9],[10,11,13],[12,13,15]], 8),13),
    Test(Input([[ 1, 5, 9],[10,11,13],[12,13,15]], 9),15),
    Test(Input([[ 1, 2, 3],[4,15,16],[7,18,19]], 5),7),
]

run_tests(tests, Solution().kthSmallest)
```
