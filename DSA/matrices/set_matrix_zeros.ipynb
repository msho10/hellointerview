Write a function to modify an m x n integer matrix matrix directly, such that if any element in the matrix is 0, its entire corresponding row and column are set to 0. This transformation should be done in place, without using any additional data structures for storage.

Input:
```
matrix = [
    [0,2,3],
    [4,5,6],
    [7,8,9]
]
```
Output:
```
[
    [0,0,0],
    [0,5,6],
    [0,8,9]
]
```
Explanation: Since the element at the first row and first column is 0, set all elements in the first row and first column to 0.

````python
from typing import List

class Solution:
    def setZeroes(self, matrix: List[List[int]]) -> None:
        if not matrix or not matrix[0]:
            return
        
        rows, cols = len(matrix), len(matrix[0])
        # 1) mark original 0's to *'s
        for r in range(rows):
            for c in range(cols):
                if matrix[r][c] == 0:
                    matrix[r][c] = '*'

        # 2) Mark entire row and col for *'s to zeros. 
        # Take care not to change cell if it is '*', otherwise, will miss some rows and cols.
        for r in range(rows):
            for c in range(cols):
                if matrix[r][c] == '*':
                    for i in range(rows):
                        if matrix[i][c] != '*':
                            matrix[i][c] = 0
                    for j in range(cols):
                        if matrix[r][j] != '*':
                            matrix[r][j] = 0
                                        
        # 3) Revert *'s back to zeros
        for r in range(rows):
            for c in range(cols):
                if matrix[r][c] == '*':
                    matrix[r][c] = 0
        
````

### Feedback

Your solution is correct for integer matrices and handles empty input safely. Marking original zeros with '*' prevents newly created zeros from affecting later decisions. However, this is not truly O(1) extra space in the strict sense: '*' is a string sentinel mixed into an integer matrix, and the nested work in step 2 can take O((mn)(m+n)) time in the worst case. For an interview, mention this tradeoff and use the first row and column as markers to achieve O(mn) time and O(1) auxiliary space. Also note that '*' would be unsafe if the matrix were allowed to contain non-integer values or if sentinel collisions were possible. Overall, your implementation passes the stated integer constraints.

````python
from typing import List

class Solution:
    def setZeroes(self, matrix: List[List[int]]) -> None:
        if not matrix or not matrix[0]:
            return

        rows, cols = len(matrix), len(matrix[0])
        # 1. mark if there is any zeros in the first row and col
        first_row_zeros = any(matrix[0][j] == 0 for j in range(cols))
        first_col_zeros = any(matrix[i][0] == 0 for i in range(rows))

        # 2. use first row and col as marker if any of the inner matrix has zeros
        for i in range(1, rows):
            for j in range(1, cols):
                if matrix[i][j] == 0:
                    matrix[i][0] = 0
                    matrix[0][j] = 0
                                        
        # 3. convert matrix[i][j] to 0 if the markers indicate that
        for i in range(1, rows):
            for j in range(1, cols):
                if matrix[i][0] == 0 or matrix[0][j] == 0:
                    matrix[i][j] = 0

        # 4. if first row and col originally have zeros, they need to be turned to zeros too
        if first_col_zeros:
            for i in range(rows):
                matrix[i][0] = 0

        if first_row_zeros:
            for j in range(cols):
                matrix[0][j] = 0     
               
````
