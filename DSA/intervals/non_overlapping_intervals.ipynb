### DESCRIPTION (inspired by Leetcode.com)

Write a function to return the minimum number of intervals that must be removed from a given array intervals, where intervals[i] consists of a starting point starti and an ending point endi, to ensure that the remaining intervals do not overlap. Intervals that only touch at their endpoints are not considered overlapping (e.g., [2,5] and [5,7] do not overlap).

##### Input:
intervals = [[1,3],[5,8],[4,10],[11,13]]

##### Output:
1

##### Explanation: 
Removing the interval [4,10] leaves all other intervals non-overlapping.

```python
import copy

class Solution:
    def nonOverlappingIntervals(self, intervals: list[list[int]]) -> int:
        if not intervals:
            return 0

        sorted_by_end = copy.deepcopy(intervals)
        sorted_by_end.sort(key=lambda x: x[1])

        count = 1
        end = sorted_by_end[0][1]
        
        for i in range(1, len(sorted_by_end)):
            if sorted_by_end[i][0] >= end:
                count += 1
                end = sorted_by_end[i][1]

        return len(sorted_by_end) - count
```

### Feedback

Correct solution — nice greedy choice by sorting on interval end time.

A couple of interview-level improvements:

copy.deepcopy(intervals) is unnecessary here and adds overhead. A shallow copy is enough if you want to avoid mutating the input.
Since you already sort a copied list, you don’t need copy at all: sorted(intervals, key=lambda x: x[1]) is cleaner.
Your logic is solid: you keep the interval with the earliest end and count compatible intervals.
Time complexity is O(n log n), space is O(n) because of the copied/sorted list.

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Test:
    input: list[list[int]]
    expected_result: int

def run_tests(tests: list[Test], func: Callable[[list[list[int]]], int]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print("Test passed for " + str(test.input))
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test([[1,3],[4,10],[11,13]],0),
    Test([[1,3],[5,8],[4,10],[11,13]],1),
    Test([[1,3],[5,8],[4,10],[11,13],[12,18]],2),
    Test([[1,4],[3,5],[4,9]],1)
]

run_tests(tests, Solution().nonOverlappingIntervals)
```
