Given three arrays, starts, ends, and profits, each representing the start time, end time, and profit of jobs respectively, your task is to schedule the jobs to maximize total profit. You can only work on one job at a time, and once a job has been started, you must finish it before starting another. The goal is to find the maximum profit that can be earned by scheduling jobs such that they do not overlap.

Input:
```
starts = [1, 3, 6, 10]
ends = [4, 5, 10, 12]
profits = [20, 20, 100, 70]
```
The optimal solution would schedule the jobs as follows:

Start the first job at time 1 and complete it by 4 for a profit of 20. Start the next job at time 6 and complete it by 10 for a profit of 100. Start the last job at time 10 and complete it by 12 for a profit of 70.

The total profit is 20 + 100 + 70 = 190

````python
from typing import List

class Solution:
    def job_scheduling(self, startTime: List[int], endTime: List[int], profit: List[int]) -> int:
        if not startTime or not endTime or not profit:
            return 0
        # dp[i] stores the max profit for jobs from 0 to i and not including jobs that are overlapping
         
        dp = [0] * len(profit)
        
        for i in range(len(profit)):
            dp[i] = profit[i]
            for j in range(i):
                if endTime[j] <= startTime[i]:
                    dp[i] = max(dp[i], dp[j]+profit[i])
                            
        return max(dp)
````

### Feedback
Your solution is correct for the tested inputs and handles empty arrays cleanly. The key assumption is that jobs are already ordered by index in increasing end time. Since the problem does not guarantee that, an unsorted input can produce an incorrect result: dp[j] only represents jobs with smaller indices, not necessarily earlier-ending jobs. Sort jobs by endTime before running this DP. Your current complexity is O(n²); after sorting, it remains O(n²), which is acceptable for small constraints but can time out for large inputs. For interview-quality performance, use binary search to find the latest compatible job, reducing the DP to O(n log n). Also, if input arrays are promised equal-length, the separate emptiness checks are sufficient; otherwise validate lengths explicitly.

````python
from bisect import bisect_right

class Solution:
    def job_scheduling(self, startTime: List[int], endTime: List[int], profits: List[int]) -> int:
        if not startTime or not endTime or not profits or len(startTime) != len(endTime) != len(profits):
            return 0

        jobs = sorted(zip(startTime, endTime, profits), key=lambda x: x[1])
        dp = [0] * (len(jobs)+1)

        for i in range(1, len(jobs)+1):
            start, end, profit = jobs[i-1]
            index = bisect_right([job[1] for job in jobs], start)
            dp[i] = max(dp[i-1], dp[index] + profit)

        return dp[-1]
````

````python
from typing import Callable

class Input:
    def __init__(self, startTime: List[int], endTime: List[int], profit: List[int]):
        self.startTime = startTime
        self.endTime = endTime
        self.profit = profit
        
class Test:  
    def __init__(self, input: Input, result: int):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[int], List[int], List[int]], int]):
    for test in tests:
        result = func(test.input.startTime, test.input.endTime, test.input.profit)
        if result == test.expected_result:
            print(f"Test passed for {test.input.startTime}")
        else:
            print(f"Test failed for {test.input.startTime}. Expected: {test.expected_result}, Actual: {result}")
````
