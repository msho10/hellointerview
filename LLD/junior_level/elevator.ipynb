## TASK

Design an elevator control system for a building. The system should handle multiple elevators, floor requests, and move elevators efficiently to service requests.

### My Questions
Q: How many elevators does the system has to support?
A: The system has 3 elevators.

Q: Are they supposed to work concurrently?
A: For this problem, the simulation runs in discrete time steps via a step() function, so concurrency is not a concern — the elevators are coordinated sequentially within each step.

Q: Can you give me an example?
A: Sure! Here's an example of how the system works:

    The building has 3 elevators (E1, E2, E3) and 10 floors (0–9).

    A user on floor 3 presses the UP button (a "hall call") — this is a request for an elevator to come pick them up.

    The system assigns one of the elevators to service that request. Let's say E1 is sent to floor 3.

    Each call to step() advances the simulation by one unit of time — so E1 might move from floor 1 → floor 2 on one step, then floor 2 → floor 3 on the next.

    Once E1 arrives at floor 3, the user boards and presses floor 7 inside the elevator (a "cabin request").

    Over subsequent step() calls, E1 travels to floor 7 and the request is fulfilled.

    Meanwhile, E2 and E3 can be servicing other requests independently during those same steps.

Q: Can a single elevator serve multiple requests at the same trip? 
A: Yes, an elevator can serve multiple requests in a single trip. For example, if E1 is heading up and has cabin requests for floors 5 and 7, it can stop at floor 5 to drop off a passenger and continue to floor 7. Similarly, it can pick up passengers along the way if there are hall calls on floors it's already passing through.

Q: Does cabin request always come after a hall call?
A: Yes, in this system a cabin request (selecting a destination floor from inside the elevator) can only happen after a passenger has boarded, which requires a hall call to have been serviced first. You can assume cabin requests are always associated with a passenger who was picked up via a hall call.

Q: After an elevator serves all requests, does it always return to the 0 floor? or does it stay at the floor where it last served?
A: After serving all requests, an elevator stays idle at the last floor it served — there is no automatic return to floor 0.

Q: How should the elevator respond when it is moving to higher floors and there is a hall call from a lower floor which the elevator has already passed?
A: If an elevator is moving upward and has already passed a floor where a hall call comes in requesting to go DOWN, that request should be left for another elevator to handle or queued for when this elevator is available — the elevator should not reverse direction mid-trip to service it. However, if another hall call comes in from a lower floor requesting to go UP, that's also already been passed, so it similarly would not be serviced on this current upward trip.

In general, an elevator does not reverse direction to pick up requests it has already passed.

Q: How does the system coordinate the 3 elevators so that they are serving requests in the most efficient way? 
A: For this problem, you can keep the dispatching logic simple — when a hall call comes in, the system should assign it to the most suitable elevator based on proximity and current direction, but the exact algorithm for "most efficient" is for you to determine in your design.

Q: Do I need to handle invalid inputs, for example, floors outside this range or hall calls for going down at floor 0?
A: Yes, invalid inputs should be rejected. This includes floor numbers outside the range 0–9, as well as logically invalid hall calls like requesting to go DOWN from floor 0 or UP from floor 9.

Q: Can an elevator be out of service, does the system need to keep track of that?
A: For this problem, elevator out-of-service states and maintenance modes are out of scope. You can assume all 3 elevators are always operational.

Q: Can some floors be unreachable? That is marked so that the elevators won't serve them?
A: For this problem, all floors in the range 0–9 are reachable by all elevators — there is no concept of restricted or unreachable floors in scope.

Q: Are UI and analytics also out of scope?
A: Yes, both UI and analytics are out of scope for this problem. You only need to focus on the core elevator control system logic.

Q: what happens when a passenger enters a floor request that is opposite to the direction of where the elevator is travelling?
A: Good question. In this system, you can assume that once a passenger is inside an elevator, their cabin request for a destination floor is always accepted and will be served — even if that floor is in the opposite direction to the elevator's current travel. The elevator would complete its current directional trip first, then reverse to service that request.
