### DESCRIPTION (inspired by Leetcode.com)
You are given a sorted array that has been rotated at an unknown pivot point, along with a target value. Develop an algorithm to locate the index of the target value in the array. If the target is not present, return -1. The algorithm should have a time complexity of O(log n).

Note:

The array was originally sorted in ascending order before being rotated.
The rotation could be at any index, including 0 (no rotation).
You may assume there are no duplicate elements in the array.
Example 1:
Input:

nums = [4,5,6,7,0,1,2], target = 0
Output: 4 (The index of 0 in the array)

Example 2:

Input:

nums = [4,5,6,7,0,1,2], target = 3
Output: -1 (3 is not in the array)

```python
class Solution:
    def search(self, nums: list[int], target: int) -> int:
        left, right = 0, len(nums)-1

        while left <= right:
            mid = (left + right) // 2
            if nums[mid] == target:
                return mid

            # there is always one half of the list in correct order (ascending)
            # check if target is in that half and adjust the pointers accordingly
            
            if nums[mid] <= nums[right]:
                if target > nums[mid] and target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
            else:
                if target < nums[mid] and target >= nums[left]:
                    right = mid - 1
                else:
                    left = mid + 1


        return -1
```

### Feedback

Correct solution—well done. Your approach identifies the sorted half using nums[mid] <= nums[right] and narrows the search in O(log n) time and O(1) space. It also correctly handles an empty array: right = -1, so the loop is skipped and -1 is returned. Because values are unique, the comparisons cleanly distinguish the halves. The boundary checks are also correct: use target > nums[mid] after confirming it is not equal, and include nums[right] (or nums[left]) appropriately. 

For interview clarity, you could briefly state the invariant that at least one half is sorted and why the target’s range determines which half to discard. Otherwise, this is an efficient and robust implementation.

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Input:
    nums: list[int]
    target: int
    
@dataclass(frozen=True)
class Test:
    input: list[Input]
    expected_result: int

def run_tests(tests: list[Test], func: Callable[[list[int], int], int]):
    for test in tests:
        result = func(test.input.nums, test.input.target)
        if result == test.expected_result:
            print("Test passed for " + str(test.input.nums))
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input([2],0),-1),
    Test(Input([3],3),0),
    Test(Input([1,3],2),-1),
    Test(Input([2,3],2),0),
    Test(Input([3,2],2),1),
    Test(Input([4,5],5),1),
    Test(Input([5,1,3],3),2),
    Test(Input([1,2,3,4],1),0),
    Test(Input([2,3,4,5],4),2),
    Test(Input([2,3,4,1],1),3),
    Test(Input([4,5,6,7,0,1,2],0),4),
    Test(Input([4,5,6,7,0,1,2],6),2),
    Test(Input([4,5,6,7,8,1,2],8),4),
    Test(Input([4,5,6,7,0,1,2],3),-1)
]

run_tests(tests, Solution().search)
```
