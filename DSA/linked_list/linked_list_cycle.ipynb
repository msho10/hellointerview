### DESCRIPTION (inspired by Leetcode.com)

Write a function that takes in a parameter head of type ListNode that is a reference to the head of a linked list. The function should return True if the linked list contains a cycle, and False otherwise, without modifying the linked list in any way.


Example 1:
                    
head -> 5 -> 4 -> 3 -> 2 -> 0 -> (back to node 3)
                    
                     
Output: true, there is a cycle between node 0 and node 3.

Example 2:

head -> 5 -> 4 -> 3 -> 2 -> 0

Output: false, there is no cycle in the linked list.

```python
#### Definition of a ListNode
class ListNode:
  def __init__(self, value=0, next=None):
    self.value = value
    self.next = next
    
class Solution_V1:
   def hasCycle(self, head: list[ListNode]) -> bool:
      if not head or not head.next:
         return False
      
      slow, fast = head, head.next.next
      
      while fast and fast.next:
         if slow == fast:
            return True
         slow = slow.next
         fast = fast.next.next
         
      return False
```

### Feedback

Your solution is correct and passes the tests. It uses Floyd’s cycle detection in O(n) time and O(1) space without modifying the list. Initializing fast = head.next.next is valid; the loop and meeting check still detect cycles, including self-cycles and two-node cycles. 

For readability, the conventional initialization slow, fast = head, head makes the invariant easier to explain: move slow one step and fast two steps, then compare them. 

Your current version compares before moving, which is also correct, but is slightly less immediately recognizable in an interview. Be prepared to explain why starting fast two links ahead does not miss any cycle.

```python
class Solution:
   def hasCycle(self, head: list[ListNode]) -> bool:
      slow, fast = head, head
      
      while fast and fast.next:
         slow = slow.next
         fast = fast.next.next
         if slow == fast:
            return True
                 
      return False
```

```python
# To construct the linked list, provide a list of integers nodes and an integer tail. 
# Each integer in nodes is used as the value of a node in the linked list, and 
# the order of the integers in the list will be the order of the nodes in the linked list.
# The tail integer is the index of the node that the last node in the linked list points 
# to form a cycle. If there is no cycle, set tail = -1.

from typing import Callable

class Test:  
    def __init__(self, nodes: list[int], tail: int, result: bool):
        self.input = nodes
        self.tail = tail
        self.expected_result = result
        
        current = ListNode()
        self.head = current
        cycle_node = None

        for i in range(len(nodes)):
            current.next = ListNode(nodes[i])
            if tail == i:
                cycle_node = current.next
            current = current.next

        current.next = cycle_node
        self.head = self.head.next

def run_tests(tests: list[Test], func: Callable[[ListNode], bool]):
    for i, test in enumerate(tests):
        result = func(test.head)
        if result == test.expected_result:
            print(f"Test passed for {test.input}, cycle index: {test.tail}")
        else:
            print(f"Test failed for {test.input}, cycle index: {test.tail}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test([], -1, False),
    Test([1], -1, False),
    Test([1, 2], -1, False),
    Test([1 ,2, 3], -1, False),
    Test([1], 0, True),
    Test([1, 2], 0, True),
    Test([1, 2, 3], 0, True),
    Test([1, 2, 3], 1, True),
    Test([1, 2, 3], 2, True),
]

run_tests(tests, Solution().hasCycle)
```
