Given an integer array nums, write a function that returns the length of the longest increasing subsequence within the array. The subsequence does not have to be contiguous, but the elements must be strictly increasing (i.e. nums[i] < nums[j] for all i < j).

Input:

nums = [8,2,4,3,6,12]

Output: 4

Explanation: The longest increasing subsequence is [2,4,6,12], which has a length of 4.

````python
from typing import List
# RECURSIVE VERSION
class Solution:
    def longest_increasing_subsequence(self, nums: List[int]) -> int:
        count = 0

        def helper(start: int, end: int) -> int:
            nonlocal count
            count += 1
            if end < start or start < 0 or end < 0 or start >= len(nums) or end>=len(nums):
                return 0
            if end == start:
                return 1
            elif end-start == 1:
                return 2 if nums[end] >= nums[start] else 1
            else:
                return helper(start, end-1) + helper(end-1, end) -1 

        result = helper(0, len(nums)-1)
        print(f"count: {count}")
        return result 
````

````python
# DP version
class Solution:
    def longest_increasing_subsequence(self, nums: List[int]) -> int:
        if not nums:
            return 0
        
        dp = [0] * len(nums)
        dp[0] = 1
                                    
        for i in range(1, len(nums)):
            dp[i] = 2 if nums[i] >= nums[i-1] else 1
            dp[i] += dp[i-1] - 1


        return dp[-1]
````

### Feedback

Your recurrence only compares adjacent elements and extends dp[i-1], so it tracks a contiguous increasing run—not a subsequence that may skip elements. In [4,10,4,3,8,9], it incorrectly carries the length from 10 into 8, even though 8 cannot follow 10. For each i, compute the best subsequence ending at i by checking every earlier j where nums[j] < nums[i]:

```
dp = [1] * len(nums)
for i in range(len(nums)):
    for j in range(i):
        if nums[j] < nums[i]:
            dp[i] = max(dp[i], dp[j] + 1)
return max(dp)
```

Your empty-input handling is correct. This dynamic-programming approach runs in O(n²) time and O(n) space.

````python
from typing import Callable

class Test:  
    def __init__(self, input: List[int], result: int):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[int]], int]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
````

````python
tests = [
    Test([], 0),
    Test([1], 1),
    Test([2,1], 1),
    Test([1,2], 2),
    Test([1,2,3], 3),
    Test([1,3,2], 2),
    Test([2,1,3], 2),
    Test([3,2,1], 1),
    Test([8,2,4,3,6,12], 4),
    Test([4,10,4,3,8,9], 3)

]

run_tests(tests, Solution().longest_increasing_subsequence)
````
