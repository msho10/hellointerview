### DESCRIPTION (inspired by Leetcode.com)

Write a function to consolidate overlapping intervals within a given array intervals, where each interval intervals[i] consists of a start time starti and an end time endi.

Two intervals are considered overlapping if they share any common time, including if one ends exactly when another begins (e.g., [1,4] and [4,5] overlap and should be merged into [1,5]).

The function should return an array of the merged intervals so that no two intervals overlap and all the intervals collectively cover all the time ranges in the original input.

Input:

intervals = [[3,5],[1,4],[7,9],[6,8]]


Output: [[1,5],[6,9]]

Explanation: The intervals [3,5] and [1,4] overlap and are merged into [1,5]. Similarly, [7,9] and [6,8] overlap and are merged into [6,9].

```python
class Solution:
    def mergeIntervals(self, intervals: list[list[int]]) -> list[list[int]]:
        if not intervals:
            return []

        sorted_intervals = sorted(intervals, key=lambda x: x[0])
        merged = []
        start = sorted_intervals[0][0]
        end = sorted_intervals[0][1]

        for i in range(1, len(sorted_intervals)):
            if sorted_intervals[i][0] > end:  # no overlap
                merged.append([start, end])
                start = sorted_intervals[i][0]
                end = sorted_intervals[i][1]
            else:
                start = min(start, sorted_intervals[i][0])
                end = max(end, sorted_intervals[i][1])

        merged.append([start, end]) 
        return merged
```

A couple of interview-level notes:

You handled the touching boundary case correctly by using > for non-overlap, so [1,4] and [4,5] merge.
Your logic is O(n log n) because of sorting, which is optimal here.
One small improvement: your else branch updates start with min(start, ...), but after sorting by start time, start never needs to change. You can simplify by only extending end. That makes the merge logic easier to reason about and less error-prone.



```python
# Same logic, but cleaner version

def mergeIntervals(intervals):
    sortedIntervals = sorted(intervals, key=lambda x: x[0])
    merged = []
        
    for interval in sortedIntervals:
        if not merged or interval[0] > merged[-1][1]:
            merged.append(interval)
        else:
            merged[-1][1] = max(interval[1], merged[-1][1])
    return merged
```

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Test:
    input: list[list[int]]
    expected_result: list[list[int]]

def run_tests(tests: list[Test], func: Callable[[list[list[int]]], list[list[int]]]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print("Test passed for " + str(test.input))
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test([],[]),
    Test([[1,5],[6,9]],[[1,5],[6,9]]),
    Test([[1,5],[6,9],[1,5],[6,9]],[[1,5],[6,9]]),
    Test([[3,5],[1,4],[7,9],[6,8]],[[1,5],[6,9]]),
]

run_tests(tests, Solution().mergeIntervals)
```
