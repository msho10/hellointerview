### DESCRIPTION (inspired by Leetcode.com)
Given a m x n integer grid image and integers sr, sc, and newColor, write a function to perform a flood fill on the image starting from the pixel image[sr][sc].

In a 'flood fill', start by changing the color of image[sr][sc] to newColor. Then, change the color of all pixels connected to image[sr][sc] from either the top, bottom, left or right that have the same color as image[sr][sc], along with all the connected pixels of those pixels, and so on.

Input:

image = [[1,0,1],[1,0,0],[0,0,1]], sr = 1, sc = 1, color = 2

Output:
[[1,2,1],[1,2,2],[2,2,1]]

The zeroes connected to the starting pixel (1, 1) are colored with the new color (2).

```python
class Solution_V1:
    def flood_fill(self, image: list[list[int]], sr: int, sc: int, color: int) -> list[list[int]]:
        org_color = image[sr][sc]
        height = len(image)
        width = len(image[0])
        visited = set()
        directions = [(0,-1),(-1,0),(1,0),(0,1)]

        def isWithinBound(r, c):
            nonlocal image, height, width
            if r < 0 or r >= height or c < 0 or c >= width:
                return False
            return True
        
        # helper function: 
        # - checks if cell has been visited. If not, adds it to visited:
        # - changes its color if cell is within bound and color matches the original color and
        # - call helper function on neighbors.

        def dfs(r, c):
            nonlocal visited, image, directions, org_color, color
            if (r,c) in visited:
                return
            visited.add((r,c))

            if isWithinBound(r, c) and image[r][c] == org_color:
                image[r][c] = color
                for dr, dc in directions:
                    dfs(r+dr, c+dc)

        dfs(sr, sc)
        return image
```

### Feedback

Your solution is correct and passed all tests. The visited set prevents cycles, and checking bounds before reading the image makes neighbor exploration safe. 

Time complexity is O(mn) in the worst case, with O(mn) auxiliary space for visited plus recursion stack. One important improvement: return immediately when org_color == color; otherwise every matching cell is still visited recursively even though no color changes, and the visited set is doing extra work. 

Also, nonlocal declarations for image, visited, etc. are unnecessary because you mutate the objects rather than rebind those variables. For interview clarity, you could simplify those parts, but your algorithm and edge-case handling are sound.

```python
class Solution:
    def flood_fill(self, image: list[list[int]], sr: int, sc: int, color: int) -> list[list[int]]:
        org_color = image[sr][sc]
        rows = len(image)
        cols = len(image[0])
        directions = [(0,-1),(-1,0),(1,0),(0,1)]

        if org_color == color:
            return image
        
        def isWithinBound(r, c):
            if r < 0 or r >= rows or c < 0 or c >= cols:
                return False
            return True
        
        # helper function: 
        # - changes its color if cell is within bound and color matches the original color and
        # - call helper function on neighbors.

        def dfs(r, c):
            if not isWithinBound(r, c) or image[r][c] != org_color:
                return
            image[r][c] = color
            
            for dr, dc in directions:
                dfs(r+dr, c+dc)

        dfs(sr, sc)
        return image
```

### Feedback

Your solution is correct and passed all tests. You correctly capture the original color, handle the important org_color == color case, mutate cells before exploring neighbors, and restrict traversal to four directions. The time complexity is O(mn) in the worst case, with O(mn) recursion-stack space. 

For interview robustness, note that Python recursion can raise RecursionError on a large, snake-shaped region; an explicit stack (iterative DFS) would avoid that limitation. 

Also, isWithinBound could be written more compactly as 0 <= r < rows and 0 <= c < cols, but that is only a readability preference.

```python
from typing import Callable

class Input:
    def __init__(self, matrix: list[list[int]], sr: int, sc: int, color: int):
        self.matrix = matrix
        self.sr = sr
        self.sc = sc
        self.color = color
        
class Test:  
    def __init__(self, input: Input, result: list[list[int]]):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[list[list[int]], int, int, int], list[list[int]]]):
    for test in tests:
        org_input = str(test.input.matrix)
        result = func(test.input.matrix, test.input.sr, test.input.sc, test.input.color)
        if result == test.expected_result:
            print(f"Test passed for {org_input}")
        else:
            print(f"Test failed for {org_input}. Expected: {test.expected_result}, Actual: {result}")
```
