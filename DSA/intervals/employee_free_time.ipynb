Write a function to find the common free time for all employees from a list called schedule. Each employee's schedule is a list of non-overlapping intervals, sorted by start time, representing the times that employee is busy (their meetings). The function should return the finite, non-zero length intervals during which every employee is free, sorted in order. Free time is the gaps that sit between busy periods once all the schedules are combined. There's no fixed workday here, so time before everyone's first meeting or after everyone's last doesn't count — that's what "finite" means.

```
Input:
schedule = [[[2,4],[7,10]],[[1,5]],[[6,9]]]

Output:
[[5,6]]
```

Explanation: The three employees collectively have only one common free time interval, which is from 5 to 6.

````python
from typing import List
import heapq

class Solution:
    def employeeFreeTime(self, schedule: List[List[List[int]]]) -> List[List[int]]:
        if not schedule or not schedule[0]:
            return []

        # 1) add all intervals to priority queue, so that we can order intervals by their start time
        heap = []
        for employee in schedule:
            for interval in employee:
                heapq.heappush(heap, interval.copy())

        # 2) merge overlapping intervals    
        merged = []
        while heap:
            next = heapq.heappop(heap)
            if not merged or next[0] > merged[-1][1]:
                merged.append(next)
            else:
                merged[-1][1] = max(merged[-1][1], next[1])
                
        # 3) find free slot between the merged intervals
        result = []
        for i in range(1, len(merged)):
            result.append([merged[i-1][1], merged[i][0]])

        return result
````

### Feedback

Your solution is correct and passed the tests. Flattening into a min-heap, merging intervals, then extracting gaps gives the right result, including overlapping meetings across employees. 

One important complexity note: because you push every interval into a heap, runtime is O(N log N) and space is O(N), where N is the total number of intervals; sorting the flattened list has the same asymptotic complexity and is often simpler. 

Your next variable shadows Python’s built-in next() function, so a name like interval would improve readability. 

Also, your initial guard checks only schedule[0]; if the first employee is empty but later employees are not, it incorrectly returns []. Use if not schedule or not any(schedule): instead.

````python
from typing import List

class Solution:
    def employeeFreeTime(self, schedule: List[List[List[int]]]) -> List[List[int]]:
        if not schedule or not any(schedule):
            return []

        # 1) flatten list and sort intervals by their start time
        flatten = [i for employee in schedule for i in employee]
        sorted_intervals = sorted(flatten, key=lambda x: x[0])

        # 2) merge overlapping intervals    
        merged = []
        for interval in sorted_intervals:
            if not merged or interval[0] > merged[-1][1]:
                merged.append(interval)
            else:
                merged[-1][1] = max(merged[-1][1], interval[1])
                
        # 3) find free slot between the merged intervals
        result = []
        for i in range(1, len(merged)):
            result.append([merged[i-1][1], merged[i][0]])

        return result
````

````python
from typing import Callable

class Test:  
    def __init__(self, input: List[List[List[int]]], result: List[List[int]]):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[List[int]]], None]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
````
