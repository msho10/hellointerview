Write a function to determine the maximum profit you can obtain from a series of stock prices given in an array prices, where prices[i] represents the stock price on the ith day. You are allowed to buy and then sell the stock once, as long as as the sell date is after the buy date. If no profit can be made, the function should return 0.

Input:
prices = [3,4,6,2,5,8]

Output: 6

Explanation: Buy on day 4 (price = 2) and sell on day 6 (price = 8), profit = 8-2 = 6.

Input:
prices = [9,7,5,3,1]

Output: 0

Explanation: Prices are in descending order, so there's no opportunity to make a profit, thus the maximum profit = 0.

```python
from typing import List

class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        if not prices:
            return 0
        
        max_profit = 0
        buy = prices[0]
        for i in range(1, len(prices)):
            if prices[i] > buy:
                max_profit = max(max_profit, prices[i] - buy)
            else:
                buy = prices[i]
                
        return max_profit
```

### Feedback

Your solution is correct and passed all tests. The greedy approach is optimal: buy stores the lowest price seen so far, and each later price is treated as a possible selling point, ensuring the sell occurs after the buy. It runs in O(n) time and O(1) space. It also correctly handles an empty list, decreasing prices, and single-element input (the loop simply does not run). For slightly clearer intent, you could use a name like min_price instead of buy, since it represents the minimum price observed rather than a specific transaction, but this is only a readability suggestion—not a correctness issue.


```python
from typing import Callable

class Test:  
    def __init__(self, input: List[int], result: int):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[int]], int]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test([1], 0),
    Test([1,1], 0),
    Test([1,3], 2),
    Test([3,4,6,2,5,8], 6),
    Test([9,7,5,3,1], 0)
]

run_tests(tests, Solution().maxProfit)
```
