Given an integer array heights representing the heights of histogram bars, write a function to find the largest rectangular area possible in a histogram, where each bar's width is 1.

Inputs:
```
heights = [2,8,5,6,2,3]
```

Output: 15

````python
from typing import List

class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        stack = [] #monotonic stack
        max_area = 0
        n = len(heights)

        # Iterate up to n to include a sentinel step that flushes out remaining elements in stack
        for i in range(n+1): 
            current_height = heights[i] if i < n else 0
            while stack and current_height < heights[stack[-1]]:
                height = heights[stack.pop()]
                left  = stack[-1] if stack else -1
                right = i - 1
                max_area = max(max_area, height * (right-left))

            stack.append(i)
            

        return max_area
````

````python
from typing import Callable

class Test:  
    def __init__(self, input: List[int], result: int):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[int]], int]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
````

````python
tests = [
    Test([], 0),
    Test([5], 5),
    Test([2,5], 5),
    Test([5,2], 5),
    Test([3,3], 6),
    Test([2,10,2,2,2,2,2], 14),
    Test([2,8,5,6,2,3], 15)
]

run_tests(tests, Solution().largestRectangleArea)
````
