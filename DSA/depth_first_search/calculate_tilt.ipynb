### DESCRIPTION (inspired by Leetcode.com)
Given the root node of a binary tree, write a recursive function to return the sum of all node tilts in the tree.

The tilt of a node is calculated as the absolute difference between the sum of all values in its left subtree and the sum of all values in its right subtree. For nodes that are missing a left child, right child, or both, treat the missing subtree as having a sum of 0.

For example, a leaf node has a tilt of 0 because both its left and right subtrees are empty (sum = 0), so the absolute difference is |0 - 0| = 0.

Example 1:

Input: [5, 1, 3]

Output: 2

The leaf nodes 1, 3 have tilts of 0 (their left and right subtrees are empty)

The root node 5 has a tilt of |1 - 3| = 2

Example 2:

Input: [4, 2, 7, 1, 3, 6, 9]

Output: 21

The leaf nodes 1, 3, 6, 9 have tilts of 0 (their left and right subtrees are empty)

Node 2 has a tilt of |1 - 3| = 2 Node 7 has a tilt of |6 - 9| = 3 Node 4 has a tilt of |6 - 22] = 16

2 + 3 + 16 = 21

```python
class TreeNode:
    def __init__(self, val: int, left: 'TreeNode' = None, right: 'TreeNode' = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def calculateTilt(self, root: TreeNode) -> int:
        tilt = 0

        def _dfs(node) -> int:   
            nonlocal tilt
            if node is None:
                return 0

            left_sum = _dfs(node.left)
            right_sum = _dfs(node.right)

            tilt += abs(left_sum - right_sum)
            return node.val + left_sum + right_sum 

        _dfs(root)  
        return tilt  
            

```

### Feedback

Correct solution—congratulations! Your postorder DFS computes each subtree’s sum after processing its children, so every node’s tilt is calculated correctly, including None roots and missing subtrees. Using nonlocal tilt cleanly accumulates the result while the recursion returns subtree sums. 

The algorithm visits each node once: O(n) time and O(h) recursion space, where h is tree height. In an interview, briefly explain why postorder traversal is necessary: a node’s tilt depends on both child subtree sums. 

For extremely deep, skewed trees, Python’s recursion limit could be a practical constraint, but recursion is explicitly appropriate for this problem.

```python
from typing import Callable

class Input:
    def __init__(self, nums: list[int]):
        self.nums = nums
        
class Test:  
    def __init__(self, input: Input, result: int):
        self.input = input
        self.expected_result = result
        self.root = self._createNodefromList(input.nums, 0)

    def _createNodefromList(self, nums: list[int], index: int) -> TreeNode:
        if index >= len(nums) or nums[index] is None:
            return None
        newNode = TreeNode(nums[index])
        newNode.left = self._createNodefromList(nums, 2*index+1)
        newNode.right = self._createNodefromList(nums, 2*index+2)
        return newNode

def run_tests(tests: list[Test], func: Callable[[TreeNode], int]):
    for test in tests:
        result = func(test.root)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums}")
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input([]),0),
    Test(Input([5]),0),
    Test(Input([5, 1]),1),
    Test(Input([5, None, 3]),3),
    Test(Input([5, 1, 3]),2),
    Test(Input([4, 2, 7, 1, 3, 6, 9]),21)
]

run_tests(tests, Solution().calculateTilt)
```
