Given a string of lowercase letters, split it into as many segments as possible so that each character appears in exactly one segment. When you concatenate all segments, they should form the original string. Return a list of integers representing the length of each segment.

Example 1
```
Input: s = "abacbcdd"

Output: [6, 2]
```
Explanation: The character 'a' appears at indices 0 and 2, 'b' at 1 and 4, 'c' at 3 and 5, 'd' at 6 and 7. To keep all occurrences of each character together, we partition into "abacbc" (length 6) and "dd" (length 2).

Example 2
```
Input: s = "eccbbbbdec"

Output: [10]
```
Explanation: All characters are interleaved such that the entire string must be one segment.

````python
from typing import List
class Solution:
    def partitionLabels(self, s: str) -> List[int]:
        # edge case
        if not s:
            return [0]
        
        # dict to store letters and the segment it belongs [start, end]
        letters = {s[0]:[0, 0]}

        # go thru each letter in the string and update the dict                
        for i in range(1, len(s)):
            if s[i] in letters:
                start = letters[s[i]][0]
                end = i
                for j in range(start, end+1):
                    letters[s[j]] = [start, end]
            else:
                letters[s[i]] = [i, i]


        # return the length for each segment
        result = []
        i = 0
        while i < len(s):
            start, end = letters[s[i]][0], letters[s[i]][1]
            result.append(end-start+1)
            i = end+1

        return result
````

### Feedback

Your solution passes and correctly expands each segment whenever a repeated character is found. 

The main issue is efficiency: for each repeat, you loop through the whole interval and rewrite dictionary entries, which can make the runtime O(n²) on inputs such as many repeated or overlapping characters. You can achieve O(n) by first recording each character’s last index, then scanning once while tracking the furthest required endpoint. Also, return [0] for an empty string is unconventional: an empty partition is usually represented by [] (and the problem may guarantee nonempty input). Your final scan is otherwise logically sound. In an interview, explain the invariant that a segment can close only when the current index reaches every character’s last occurrence.


````python
from typing import Callable

class Test:  
    def __init__(self, input: str, result: List[int]):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[str], List[int]]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
````

````python
tests = [
    Test("", [0]),
    Test("a", [1]),
    Test("aa", [2]),
    Test("ab", [1, 1]),
    Test("aba", [3]),
    Test("abacbcdd", [6, 2]),
    Test("eccbbbbdec", [10])
]

run_tests(tests, Solution().partitionLabels)
````
