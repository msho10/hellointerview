You are provided with a string s and a set of words called wordDict. Write a function to determine whether s can be broken down into a sequence of one or more words from wordDict, where each word can appear more than once and there are no spaces in s. If s can be segmented in such a way, return true; otherwise, return false.

Input:

s = "catsandog", wordDict = ["cats","dog","sand","and","cat"]

Output: false

Explanation: There is no valid segmentation of "catsandog" into dictionary words from wordDict.

Input:

s = "hellointerview", wordDict = ["hello","interview"]

Output: true

Explanation: Return true because "hellointerview" can be segmented as "hello" and "interview".

Note that you are allowed to reuse a dictionary word.

```python
from typing import List
# Recursive solution
class Solution:
    def word_break(self, s: str, wordDict: List[str]) -> bool:
        if not s or not wordDict:
            return False

        for word in wordDict:
            if word == s:
                return True
            if word == s[:len(word)]:
                if self.word_break(s[len(word):], wordDict):  # if this word fails, keep trying, there may be words in the rest of the dict that works
                    return True

        return False
```

### Feedback

Your recursive solution is correct and handles backtracking properly: if one matching prefix fails, the loop tries other dictionary words. 

However, it has exponential worst-case time because the same suffix is recomputed along many recursive paths, and recursion depth can reach len(s), risking Python's recursion limit. Add memoization keyed by the remaining suffix (or index) to make it roughly O(n · m · L), where m is the dictionary size and L is the maximum word length. Also, if not s returns False; that matches the usual nonempty-input expectation, but conceptually an empty string is often considered segmentable, so confirm the problem's convention. Converting wordDict to a set can make exact membership checks faster, though prefix iteration remains the main cost.

```python
# DP Solution

class Solution:
    def word_break(self, s: str, wordDict: List[str]) -> bool:
        dp = [False] * (len(s)+1)
        dp[0] = True

        for i in range(1, len(s)+1):
            for word in wordDict:
                if i >= len(word) and dp[i-len(word)]:
                    sub = s[i-len(word):i]
                    if word == sub:
                        dp[i] = True
                        break
                    
        return dp[-1]
```

```python
from typing import Callable

class Input:
    def __init__(self, s:str, wordDict: List[str]):
        self.s = s
        self.wordDict = wordDict
        
class Test:  
    def __init__(self, input: Input, result: bool):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[str, List[str]], bool]):
    for test in tests:
        result = func(test.input.s, test.input.wordDict)
        if result == test.expected_result:
            print(f"Test passed for {test.input.s}")
        else:
            print(f"Test failed for {test.input.s}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input("", ["cats", "dogs"]), True),
    Test(Input("empty", []), False),
    Test(Input("", []), True),
    Test(Input("bats", ["bats"]), True),
    Test(Input("batnos", ["bat"]), False),
    Test(Input("bat", ["at"]), False),
    Test(Input("at", ["bat"]), False),
    Test(Input("babat", ["bat"]), False),
    Test(Input("batss", ["bat"]), False),
    Test(Input("catsandog", ["cats","dog","sand","and","cat"]), False),
    Test(Input("catsanddog", ["cat","dog","sand","and","cats"]), True),
    Test(Input("catsandcats", ["cat","dog","sand","and","cats"]), True),        
    Test(Input("catsanddog", ["cats","dogs","sand","and","cat"]), False),       
    Test(Input("hellointerview", ["hello","interview"]), True),
    Test(Input("hellowinterview", ["hello","interview"]), False),
]

run_tests(tests, Solution().word_break)
```
