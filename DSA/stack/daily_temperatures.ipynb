### DESCRIPTION (inspired by Leetcode.com)
Given an integer array temps representing daily temperatures, write a function to calculate the number of days one has to wait for a warmer temperature after each given day. The function should return an array answer where answer[i] represents the wait time for a warmer day after the ith day. If no warmer day is expected in the future, set answer[i] to 0.

Inputs:

temps = [65, 70, 68, 60, 55, 75, 80, 74]

Output:

[1,4,3,2,1,1,0,0]

```python
# Intention: for each element, find the index for the first element encountered afterwards 
# that has a higher value.
# Approach: use a monotonic stack to keep track of indices for elements that has larger values 
# than the current iterate element. If the current element is larger than the ones in the stack, 
# we will pop the indices from the stack, calculate the difference between the current element index 
# and the popped index and update the answer list.

class Solution:
    def dailyTemperatures(self, temps: list[int]) -> list[int]:
        stack = []
        answer = [0 for _ in range(len(temps))]

        for i in range(len(temps)):
            while stack and temps[i] > temps[stack[-1]]:
                idx = stack.pop()
                answer[idx] = i-idx
            stack.append(i)

        return answer
```

### Feedback

Excellent solution. Your monotonic decreasing stack stores indices whose warmer day has not yet appeared. When a warmer temperature arrives, popping each cooler index and assigning i - idx is correct; indices remaining in the stack appropriately retain the initialized 0. 

The algorithm runs in O(n) time because each index is pushed and popped at most once, with O(n) auxiliary space. You could use [0] * len(temps) for slightly simpler initialization, but your current version is fully correct. It also handles empty input and equal temperatures correctly, since equal values do not count as warmer.

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Test:
    input: list[list[int]]
    expected_result: list[list[int]]

def run_tests(tests: list[Test], func: Callable[[list[list[int]]], list[list[int]]]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print("Test passed for " + str(test.input))
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests =[
    Test([],[]),
    Test([88],[0]),
    Test([56, 98], [1,0]),
    Test([56, 50], [0,0]),
    Test([50, 50], [0,0]),
    Test([65, 70, 68, 60, 55, 75, 80, 74], [1,4,3,2,1,1,0,0])
]

run_tests(tests, Solution().dailyTemperatures)
```
