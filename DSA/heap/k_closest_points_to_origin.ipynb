### DESCRIPTION (inspired by Leetcode.com)
Given a list of points in the form [[x1, y1], [x2, y2], ... [xn, yn]] and an integer k, find the k closest points to the origin (0, 0) on the 2D plane.

The distance between two points (x, y) and (a, b) is calculated using the formula:

√(x1 - a2)2 + (y1 - b2)2
Return the k closest points in any order.

Example 1:

Inputs:

points = [[3,4],[2,2],[1,1],[0,0],[5,5]]
k = 3
Output:

[[2,2],[1,1],[0,0]]
Also valid:

[[2,2],[0,0],[1,1]]
[[1,1],[0,0],[2,2]]
[[1,1],[2,2],[0,0]]
...
[[0,0],[1,1],[2,2]]

```python
import heapq
import math

class Solution_V1:
    def kClosest(self, points: list[list[int]], k: int) -> list[list[int]]:
        if not points or len(points) == 0 or k == 0:
            return []
        if len(points) == 1 or len(points) == k:
            return points

        max_heap = []
        for point in points[:k]:
            distance = 0 - math.sqrt(point[0]**2 + point[1]**2)
            heapq.heappush(max_heap, (distance, point))

        for point in points[k:]:
            distance = 0 - math.sqrt(point[0]**2 + point[1]**2)
            if distance > max_heap[0][0]:
                heapq.heapreplace(max_heap, (distance, point))
                
        return [point for distance, point in max_heap]
```

### Feedback

Your solution is correct and efficiently maintains a heap of size k, giving O(n log k) time and O(k) extra space. Nice use of negative distances to simulate a max-heap. 

A few improvements: not points already covers len(points) == 0, and len(points) == 1 is unnecessary because the general logic handles it. 

More importantly, use squared distances (x*x + y*y) instead of math.sqrt(...): square root is unnecessary for comparisons and avoids floating-point work. 

You should also ensure heapq and math are imported. The returned order is unspecified, which is valid unless the interviewer requires sorting. For readability, compare positive squared distances and negate only the heap key.

```python
import heapq
import math

class Solution:
    def kClosest(self, points: list[list[int]], k: int) -> list[list[int]]:
        if not points or len(points) == 0 or k == 0:
            return []
        if len(points) == 1 or len(points) == k:
            return points

        max_heap = []
        for point in points[:k]:
            distance = point[0]**2 + point[1]**2
            heapq.heappush(max_heap, (0 - distance, point))

        for point in points[k:]:
            distance = point[0]**2 + point[1]**2
            if distance < 0 - max_heap[0][0]:
                heapq.heapreplace(max_heap, (0 - distance, point))
                
        return [point for distance, point in max_heap]
```

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Input:
    points: list[list[int]]
    k: int
    
@dataclass(frozen=True)
class Test:
    input: Input
    expected_result: int

def run_tests(tests: list[Test], func: Callable[[list[list[int]], int], list[list[int]]]):
    for test in tests:
        result = func(test.input.points, test.input.k)
        if verify(result, test.expected_result):
            print("Test passed for " + str(test.input.points))
        else:
            print(f"Test failed for {test.input.points}. Expected: {test.expected_result}, Actual: {result}")

def verify(actual_result: list[list[int]], expected_result: list[list[int]]) -> bool:
    if len(actual_result) != len(expected_result):
        return False
    
    for item in expected_result:
        if item not in actual_result:
            return False

    return True
            
```

```python
tests = [
    Test(Input([[0,0]], k=0), []),
    Test(Input([[0,0]], k=1), [[0,0]]),
    Test(Input([[0,0],[1,1]], k=1), [[0,0]]),
    Test(Input([[0,0],[1,1]], k=2), [[0,0],[1,1]]),
    Test(Input([[0,5],[0,-2]], k=1), [[0,-2]]), 
    Test(Input([[5,0],[-3,0]], k=1), [[-3,0]]),   
    Test(Input([[3,4],[2,2],[1,1],[0,0],[5,5]],k=3), [[2,2],[1,1],[0,0]]),
]

run_tests(tests, Solution().kClosest)
```
