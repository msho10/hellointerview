### DESCRIPTION (inspired by Leetcode.com)
medium

You are a renowned painter who is given a task to paint n houses in a row. You can paint each house with one of three colors: Red, Blue, or Green. The cost of painting each house with each color is different and given in a 2D array costs:

```
costs[i][0] = cost of painting house i Red
costs[i][1] = cost of painting house i Blue
costs[i][2] = cost of painting house i Green
```
No two neighboring houses can have the same color. Return the minimum cost to paint all houses.

Constraints:
```
1 ≤ n ≤ 100
costs[i].length == 3
0 ≤ costs[i][j] ≤ 1000
```

Example 1

Input:
```
costs = [[8, 4, 15], [10, 7, 3], [6, 9, 12]]
```
Output: 13

Explanation:
```
House 0: Blue (cost = 4)
House 1: Green (cost = 3)
House 2: Red (cost = 6)
Total = 4 + 3 + 6 = 13
```

Example 2

Input:
```
costs = [[5, 8, 6], [19, 14, 13], [7, 5, 12], [14, 5, 9]]
```
Output: 30


Explanation: Red(5) → Green(13) → Red(7) → Blue(5) = 30

````python
from typing import List

# RECURSIVE USING MEMOIZATION 
class Solution:
    def min_cost(self, costs: List[List[int]]) -> int:
        if not costs or not costs[0]:
            return 0
        memo = {0 : (costs[0][0],costs[0][1],costs[0][2])}

        # helper function to return the min cost for each of the colors until house i
        def helper(i):
            if i < 0:
                return (0, 0, 0)
            if i in memo:
                return memo[i]

            prev_cost = helper(i-1)
            color_0_min = min(costs[i][0] + prev_cost[1], costs[i][0] + prev_cost[2])
            color_1_min = min(costs[i][1] + prev_cost[0], costs[i][1] + prev_cost[2])
            color_2_min = min(costs[i][2] + prev_cost[0], costs[i][2] + prev_cost[1])
            memo[i] = (color_0_min, color_1_min, color_2_min)
            
            return memo[i]

        min_costs = helper(len(costs)-1)
        return min(min_costs) 
````

````python
# USING DP
class Solution:
    def min_cost(self, costs: List[List[int]]) -> int:
        if not costs or not costs[0]:
            return 0
        dp = [[0,0,0]] * (len(costs)+1)
        dp[1] = costs[0]
        
        for i in range(1, len(costs)+1):
            prev_cost = dp[i-1]
            current_cost = costs[i-1]
            color_0_min = min(current_cost[0] + prev_cost[1], current_cost[0] + prev_cost[2])
            color_1_min = min(current_cost[1] + prev_cost[0], current_cost[1] + prev_cost[2])
            color_2_min = min(current_cost[2] + prev_cost[0], current_cost[2] + prev_cost[1])
            dp[i] = [color_0_min, color_1_min, color_2_min]
            
        return min(dp[-1]) 
````

### Feedback

Your dynamic programming recurrence is correct, and passing all tests confirms the boundary handling works for empty input and normal cases. The state dp[i][color] represents the minimum cost through the first i houses ending in that color, giving O(n) time and O(n) space. You can reduce space to O(1), since each row only depends on the previous row; this is useful to mention in an interview. Also, [[0,0,0]] * (len(costs)+1) can be misleading because it creates repeated references, though your subsequent assignments prevent a bug here. Initializing with [None] * ... or using three rolling values would make the intent clearer. Note that not costs[0] assumes valid, consistently shaped rows.

````python
from typing import Callable

class Test:  
    def __init__(self, input: List[List[int]], result: int):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[List[int]]], int]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
````
