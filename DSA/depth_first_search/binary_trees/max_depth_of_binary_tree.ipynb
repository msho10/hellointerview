### DESCRIPTION (inspired by Leetcode.com)
Given the root of a binary tree, write a recursive function to find its maximum depth, where maximum depth is defined as the number of nodes along the longest path from the root node down to a leaf node.

Example 1:

Input:

[4, 2, 7, 1, null, 6, 9, null, 8, null, null, null, null, null, null]

Output: 4 (The 4 nodes along the longest path are shown in bold)

```python
class TreeNode:
    def __init__(self, val: int, left: 'TreeNode' = None, right: 'TreeNode' = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def maxDepth(self, root: TreeNode) -> int:
        if not root:
            return 0
        else:
            left_depth = self.maxDepth(root.left)
            right_depth = self.maxDepth(root.right)
            return 1 + max(left_depth, right_depth)
```

### Feedback

Your solution is correct and clean. The base case returns 0 for an empty tree, and each non-empty node contributes 1 plus the deeper subtree, so it handles balanced, skewed, and single-node trees correctly. 

Complexity is O(n) time, since each node is visited once, and O(h) space for the recursion stack, where h is the tree height. 

For an interview, briefly explain that the recursion computes subtree depths bottom-up. 

One practical Python consideration: an extremely skewed tree can exceed Python’s recursion limit; an iterative BFS/DFS would avoid that, though your recursive approach is the standard expected solution here.

```python
from typing import Callable

class Test:  
    def __init__(self, nums: list[int], result: int):
        self.input = nums
        self.expected_result = result
        self.root = self._createNodefromList(nums, 0)

    def _createNodefromList(self, nums: list[int], index: int) -> TreeNode:
        if index >= len(nums) or not nums[index]:
            return None
        newNode = TreeNode(nums[index])
        newNode.left = self._createNodefromList(nums, 2*index+1)
        newNode.right = self._createNodefromList(nums, 2*index+2)
        return newNode

def run_tests(tests: list[Test], func: Callable[[TreeNode], int]):
    for test in tests:
        result = func(test.root)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test([None],0),
    Test([1], 1),
    Test([1,2], 2),
    Test([1,2,3], 2),
    Test([1,2,3,4], 3),
    Test([1,2,3,None,4], 3),
    Test([1,2,3,None,None,4], 3),
    Test([1,2,3,None,None,None,4], 3),
    Test([4, 2, 7, 1, None, 6, 9, None, 8, None, None, None, None, None, None],4)
]

run_tests(tests, Solution().maxDepth)
```
