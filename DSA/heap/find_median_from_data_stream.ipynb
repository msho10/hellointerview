A live analytics dashboard starts with nums. New readings arrive in order in adds. After each incoming value is inserted, return the current median of all readings seen so far. If the count is even, the median is the average of the two middle values.

Example 1:
```
Inputs:

nums = [5, 2, 8]
adds = [3, 10, 4]

Output:[4, 5, 4.5]
```

Median is the middle value of a collection of numbers if its size is odd or average between the 2 middle numbers if size is even.
I can separate the numbers into 2 groups, a lower half for the smaller values and a upper half for the larger values. 
The lower half will be a max heap, so that the largest number will be always on top. 
The upper half will be a min heap, so that the smallest number will be on top.
Median can be obtained or calculated by the tops of the 2 heaps.
The invariant is after each insert operation, the lower half should have same or at most 1 number more than the upper half. 
Median will be the top of the lower heap if its length is greater than the upper heap. If both heaps have the same length, median is the average of the top of both heaps.


````python
from typing import List
import heapq

class medianFinder():
    def __init__(self):
        self._lo = []
        self._hi = []

    def insert(self, num: int) -> float:
        heapq.heappush(self._lo, -num)
        heapq.heappush(self._hi, -heapq.heappop(self._lo))

        if len(self._hi) > len(self._lo):
            heapq.heappush(self._lo, -heapq.heappop(self._hi))

        if len(self._hi) == len(self._lo):    
            return (-self._lo[0] + self._hi[0]) / 2
        else:
            return -self._lo[0]    

    
class Solution:
    def medianStream(self, nums: List[int], adds: List[int]) -> List[float]:
        mf = medianFinder()
        for n in nums:
            mf.insert(n)

        return [mf.insert(a) for a in adds]
````

````python
from typing import Callable

class Input:
    def __init__(self, nums: List[int], adds: List[int]):
        self.nums = nums
        self.adds = adds

class Test:  
    def __init__(self, input: Input, result: List[float]):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[int], List[int]], List[float]]):
    for test in tests:
        result = func(test.input.nums, test.input.adds)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums, test.input.adds}")
        else:
            print(f"Test failed for {test.input.nums, test.input.adds}. Expected: {test.expected_result}, Actual: {result}")
````
