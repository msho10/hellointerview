medium

Write a function that returns the total number of contiguous subarrays within a given integer array whose elements sum up to a target K.

Example 1: 
```
Input:
nums = [3, 4, 7, 2, -3, 1, 4, 2]
k = 7
```
Output: 4

Explanation: 

The subarrays that sum to 7 are:
[3, 4], [7], [7, 2, -3, 1], [1, 4, 2]

Example 2: 
```
Input:
nums = [1, -1, 0]
k = 0
```
Output: 3

Explanation: 
The subarrays that sum to 0 are:
[-1, 1], [0], [1, -1, 0]

````python
from typing import List

class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        n = len(nums)
        prefix = [0] * (n+1)
        for i in range(1, n+1):
            prefix[i] = prefix[i-1] + nums[i-1]

        count = 0
        for i in range(n+1):
            for j in range(i):
                if prefix[i] - k == prefix[j]:
                    count += 1

        return count        
````

### Feedback

Your solution is correct and handles duplicates, negative values, k = 0, and empty input properly. The prefix-sum indexing is also sound: requiring j < i guarantees non-empty subarrays. 

The main interview concern is efficiency: the nested loops take O(n²) time, which can be too slow for large arrays, while the prefix array uses O(n) extra space. You can improve this to O(n) time by storing how many times each prefix sum has appeared. For each current sum s, add the count of s - k seen previously, then record s. This is especially important because repeated prefix sums represent multiple valid starting positions.

````python
from typing import List

class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        count = 0
        current_sum = 0
        prefix_count = {0: 1}

        for n in nums:
            current_sum += n
            if current_sum - k in prefix_count:
                count += prefix_count[current_sum-k]
                
            prefix_count[current_sum] = prefix_count.get(current_sum, 0) + 1

        return count  
````

````python
from typing import Callable

class Input:
    def __init__(self, nums: List[int], k: int):
        self.nums = nums
        self.k = k
        
class Test:  
    def __init__(self, input: Input, result: int):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[List[int], int], int]):
    for test in tests:
        result = func(test.input.nums, test.input.k)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums}")
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
````

````python
tests = [
    Test(Input([], 4), 0),
    Test(Input([0], 4), 0),
    Test(Input([1], 1), 1),
    Test(Input([1, 5], 2), 0),
    Test(Input([1, 2], 2), 1),
    Test(Input([1, 3], 4), 1),
    Test(Input([3, 4, 7, 2, -3, 1, 4, 2], 7), 4),
    Test(Input([1, -1, 0], 0), 3)
]

run_tests(tests, Solution().subarraySum)
````
