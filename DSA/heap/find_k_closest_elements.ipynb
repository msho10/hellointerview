### DESCRIPTION (inspired by Leetcode.com)
Given a sorted array nums, a target value target, and an integer k, find the k closest elements to target in the array, where "closest" is the absolute difference between each element and target. Return these elements in array, sorted in ascending order.

Example 1:

Inputs:

nums = [-1, 0, 1, 4, 6]
target = 1
k = 3

Output:

[-1, 0, 1]

Explanation: -1 is 2 away from 1, 0 is 1 away from 1, and 1 is 0 away from 1. All other elements are more than 2 away. Since we need to return the elements in ascending order, the answer is [-1, 0, 1]

Example 2:

Inputs:

nums = [5, 6, 7, 8, 9]
target = 10
k = 2

Output:

[8, 9]

```python
import heapq

class Solution:
    def kClosest(self, nums: list[int], k: int, target: int) -> list[int]:
        if k==0 or not nums:
            return []
        if k>=len(nums):
            return nums
         
        max_heap = []
        for i in range(len(nums)):
            diff = abs(target - nums[i])
            if i < k:
                heapq.heappush(max_heap, (-diff, nums[i]))
            else:
                if diff < -max_heap[0][0]:
                    heapq.heapreplace(max_heap, (-diff, nums[i]))

        # now max_heap contains tuples of k closest numbers, 
        # we need to return the numbers in ascending order
        return sorted([num for diff, num in max_heap])


```

Your solution is correct and passed all tests. The max-heap approach maintains the k smallest distances in O(n log k) time and O(k) space, then sorts the result in O(k log k) as required. 

Your tie behavior is valid: when distances are equal, you keep the earlier element already in the heap, which is acceptable unless the problem specifies a particular tie-break rule.

```python
from dataclasses import dataclass
from typing import Callable

@dataclass(frozen=True)
class Input:
    nums: list[int]
    target: int
    k: int
    
@dataclass(frozen=True)
class Test:
    input: Input
    expected_result: list[int]

def run_tests(tests: list[Test], func: Callable[[list[int], int, int], list[int]]):
    for test in tests:
        result = func(test.input.nums, test.input.k, test.input.target)
        if result == test.expected_result:
            print("Test passed for " + str(test.input.nums))
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")

            
```
