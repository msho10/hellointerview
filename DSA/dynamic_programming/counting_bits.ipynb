Write a function that, given an integer n, returns an array dp of size n + 1, where dp[i] stores the count of '1' bits in the binary form of i.

Input:

n = 6


Output: [0,1,1,2,1,2,2]

Explanation:
```
0 --> 0
1 --> 1
2 --> 10
3 --> 11
4 --> 100
5 --> 101
6 --> 110
```

````python
from typing import List

class Solution:
    def count_bits(self, n: int) -> List[int]:
        if n == 0:
            return [0]
        
        counts = [0] * (n+1)
        
        for i in range(1, n+1):
            # both of these work:

            #counts[i] = counts[i//2] + i%2  
            counts[i] = counts[i >> 1] + (i & 1)

        return counts
````

````python
from typing import Callable

class Test:  
    def __init__(self, input: int, result: List[int]):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[int], List[int]]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
````

````python
tests = [
    Test(0, [0]),
    Test(1, [0,1]),
    Test(6, [0,1,1,2,1,2,2])
]

run_tests(tests, Solution().count_bits)
````
