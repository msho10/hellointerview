You are given a robot that starts at the top-left corner of a grid with dimensions m x n. The robot can only move either down or right at any point in time. The goal is for the robot to reach the bottom-right corner of the grid.

Given the dimensions of the board m and n, write a function to return the number of unique paths the robot can take to reach the bottom-right corner.

Example 1

Input:
```
m = 2
n = 3
```
Output: 3

Explanation: The robot starts at the top-left corner and can move right or down. There are 3 unique ways to reach the bottom-right corner of a 2 x 3 grid.

Example 2

Input:
```
m = 3
n = 7
```
Output: 28

````python
# BASIC RECURSIVE SOLUTION
class Solution:
    def unique_paths(self, m: int, n: int) -> int:
        count = 0
        big_o = 0

        def dfs(r: int, c: int):
            nonlocal count, m, n, big_o
            big_o += 1 
            if r == m-1 and c == n-1:
                count += 1
                return
            if r < m-1:
                dfs(r+1, c)
            if c < n-1:
                dfs(r, c+1)

        dfs(0, 0)
        print(f"big-O: {big_o}")
        return count
````

````python
# DP with recursive and memoization
class Solution:
    def unique_paths(self, m: int, n: int) -> int:
        big_o = 0
        if m <= 0 or n <= 0:
            return 0
        memo = {}
        
        def helper(r: int, c: int) -> int:
            nonlocal big_o
            big_o += 1
            if r == 1 or c == 1 :
                return 1 
            
            if (r, c) in memo:
                return memo[(r, c)]
            
            memo[(r, c)] = helper(r-1, c) + helper(r, c-1)
            return memo[(r, c)]
        
        count = helper(m, n)
        print(f"big-O: {big_o}")
        return count
````

````python
# DP using bottom-up and iteration
class Solution:
    def unique_paths(self, m: int, n: int) -> int:
        big_o = 0
        if m <= 0 or n <= 0:
            return 0
        
        dp = [[0] * (n+1) for _ in range(m+1)]
        for r in range(1, m+1):
            dp[r][1] = 1
        for c in range(1, n+1):
            dp[1][c] = 1
        
        for r in range(2, m+1):
            for c in range(2, n+1):
                big_o += 1
                dp[r][c] = dp[r][c-1] + dp[r-1][c]

        print(f"big-O: {big_o}")
        return dp[m][n]
````
