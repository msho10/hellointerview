medium

Given an integer n, write a function to return all well-formed (valid) expressions that can be made using n pairs of parentheses.

Example 1:

Input:
n = 3

Output:

["((()))","(()())","(())()","()(())","()()()"]

Example 2:

Input:
n = 2

Output:
["()()", "(())"]

```python
from typing import List

class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        if n < 1:
            return [""]
        
        result = set()
        def dfs(level: int, parenthesis: str):
            if level == n:
                result.add(parenthesis)
                return
                
            dfs(level+1, f"{parenthesis}()")
            dfs(level+1, f"(){parenthesis}")
            dfs(level+1, f"({parenthesis})")

        dfs(1, "()")
            
        return list(result)
```

### Feedback

Your recursion only inserts "()" around or beside an existing string. That misses valid structures such as (())(()), because constructing valid strings by local insertion this way is not complete; the set removes duplicates but cannot recover omitted results. Build strings incrementally while tracking counts of opening and closing parentheses.


```python
from typing import List

class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        result = []
        
        def dfs(parentheses: str, open: int, close: int):
            if len(parentheses) == 2*n:
                result.append(parentheses)
                return
                
            if open < n:
                dfs(parentheses + "(", open+1, close)
            if close < open:
                dfs(parentheses + ")", open, close+1)

        dfs("", 0, 0)
            
        return result
```

```python
from typing import Callable

class Test:  
    def __init__(self, n: int, result: List[str]):
        self.n = n
        self.expected_result = result

    def verify(self, actual_result: List[str]) -> False:
        verify_result = True

        if self.expected_result is None:
            return actual_result is None
         
        for item in self.expected_result:
            if item not in actual_result:
                print(f"missing item: {item}")
                verify_result = False

        # in case actual_result contains more:
        if len(self.expected_result) < len(actual_result):
            for item in actual_result:
                if item not in self.expected_result:
                    print(f"item not expected: {item}")
                    verify_result = False
                
        return verify_result
        
def run_tests(tests: list[Test], func: Callable[[int], List[str]]):
    for test in tests:
        result = func(test.n)
        if test.verify(result):
            print(f"Test passed for {test.n}")
        else:
            print(f"Test failed for {test.n}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(0, [""]),
    Test(1, ["()"]),
    Test(2, [ "(())","()()"]),
    Test(3, ["((()))","(()())","(())()","()(())","()()()"]),
    Test(4, ["(((())))","((()()))","((())())","((()))()","(()(()))","(()()())","(()())()","(())(())","(())()()","()((()))","()(()())","()(())()","()()(())","()()()()"])
]

run_tests(tests, Solution().generateParenthesis)
```
