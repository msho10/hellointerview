### DESCRIPTION (inspired by [Leetcode.com](https://leetcode.com/problems/swap-nodes-in-pairs))

Given a reference head of type ListNode that is the head of a singly linked list, write a function to swap every two adjacent nodes and return its head.

You must solve the problem without modifying the values in the list's nodes (i.e., only nodes themselves may be changed.)

**Example 1:** 

input:

head -> 5 -> 4 -> 3 -> 2 -> 1

output:

 head -> 4 -> 5 -> 2 -> 3 -> 1

Explanation: 5 and 4 are swapped, 3 and 2 are swapped, and 1 is left alone.

**Example 2:** 

input:

head -> 1 -> 2 -> 3 -> 4

output:

 head -> 2 -> 1 -> 4 -> 3

Explanation: 1 and 2 are swapped, 3 and 4 are swapped.

```python
class ListNode:
    def __init__(self, val: int = 0, next: 'ListNode' = None):
        self.val = val
        self.next = next

    # convert node list to [], for easy display
    def toList(self) -> list[int]:
        current = self
        result = []
        while current:
            result.append(current.val)
            current = current.next

        return result
    
class Solution:
    def swapPairs(self, head: ListNode) -> ListNode:
        dummy = ListNode(0, head)
        prev = dummy
        slow = fast = head
        
        while slow and slow.next:
            fast = slow.next.next
            prev.next = slow.next
            slow.next.next = slow
            slow.next = fast

            prev = slow
            slow = fast

        return dummy.next
```

### Feedback

Your solution is correct and efficiently swaps nodes in place, with O(n) time and O(1) extra space. The dummy node handles an empty list and ensures the new head is returned correctly. fast is used to save the node after the current pair before rewiring, and prev is advanced to the first node after the swap. 

The unused initial assignment slow = fast = head is harmless, but fast need not be initialized separately because it is assigned before use in the loop. 

For interview clarity, consider naming slow/fast as first/after_pair, since this is pointer rewiring rather than fast/slow traversal.

```python
from typing import Callable
class Test:  
    def __init__(self, nodes: list[int], result: list[int]):
        self.input = nodes
        self.expected_result = result
        
        current = ListNode()
        self.head = current
        
        for i in range(len(nodes)):
            current.next = ListNode(nodes[i])
            current = current.next

        self.head = self.head.next

def run_tests(tests: list[Test], func: Callable[[ListNode], None]):
    for test in tests:
        actual_result = []
        new_head = func(test.head)
        if new_head:
            actual_result = new_head.toList()
            
        if  actual_result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {actual_result}")
```

```python
tests = [
    Test([],[]),
    Test([0], [0]),
    Test([0,1], [1,0]),
    Test([0,1,2], [1,0,2]),
    Test([0,1,2,3], [1,0,3,2]),
    Test([0,1,2,3,4], [1,0,3,2,4]),
]

run_tests(tests, Solution().swapPairs)
```
