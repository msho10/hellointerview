### DESCRIPTION (inspired by Leetcode.com)
medium
You are given a string s containing only digits. Write a function to return the number of ways to decode using the following mapping:
```
'1' -> "A"
'2' -> "B"
'3' -> "C"
...
'26' -> "Z"
```

There may be multiple ways to decode a string. For example, "14" can be decoded as "AD" or "N".

Input:
s = 101

Output: 1

Explanation: The only way to decode it is "JA". "01" cannot be decoded into "A" as "1" and "01" are different.

````python
# Using backtracking

class Solution:
    def num_decodings(self, s: str) -> int:
        count = 0

        def dfs(s: str):
            nonlocal count
            if not s:
                count += 1
                return
            if s[0] == '0':
                return
            if len(s) > 1 and int(s[:2]) <= 26:
                dfs(s[2:])

            dfs(s[1:])

        dfs(s)
        return count


 
````

### Feedback
Your DFS is correct: it rejects leading/embedded 0s appropriately and explores valid one- and two-digit choices. It passes all tests, but it recomputes the same suffixes many times, giving exponential time complexity—roughly O(2ⁿ)—and recursion depth O(n). 

For interview-quality scalability, memoize by position rather than passing sliced strings; slicing also creates O(n) copies per call. For example, cache dfs(i) and inspect s[i]/s[i:i+2]. Alternatively, bottom-up DP achieves O(n) time and O(n) space (or O(1) space). Your count closure works, but returning the number of ways from each recursive call is generally easier to reason about and avoids shared mutable state.


````python
# Using dynamic programming, memoization

class Solution:
    def num_decodings(self, s: str) -> int:
        if not s or s[0] == "0":
            return 0

        n = len(s)
        dp = [0] * (n+1)
        dp[0], dp[1] = 1, 1

        for i in range(2, n+1):
            if s[i-1] != "0":  #s[i] counts as valid, only if the preceding is not a zero
                dp[i] = dp[i-1]

            if 10 <= int(s[i-2:i]) <= 26:
                dp[i] += dp[i-2]
                
        return dp[-1]


 
````

````python
from typing import Callable

class Test:  
    def __init__(self, input: str, result: int):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[str], int]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
````

````python
tests = [
    Test("2", 1),
    Test("23", 2),
    Test("101", 1),
    Test("123", 3),
    Test("1122", 5),
]

run_tests(tests, Solution().num_decodings)
````
