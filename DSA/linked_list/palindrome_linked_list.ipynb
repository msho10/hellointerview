### DESCRIPTION (inspired by Leetcode.com)

Given a reference of type ListNode which is the head of a singly linked list, write a function to determine if the linked list is a palindrome.

A linked list is a palindrome if the values of the nodes are the same when read from left-to-right and right-to-left. An empty list is considered a palindrome.

Example 1:

head -> 5 -> 4 -> 3 -> 4 -> 5
Output: True

left-to-right:  5 -> 4 -> 3 -> 4 -> 5
right-to-left: 5 -> 4 -> 3 -> 4 -> 5   

Example 2:

head -> 5 -> 4 -> 3
Output: False

left-to-right:  5 -> 4 -> 3
right-to-left: 3 -> 4 -> 5

```python
class ListNode:
    def __init__(self, val: int = 0, next: 'ListNode' = None):
        self.val = val
        self.next = next

class Solution:
    def isPalindrome(self, head: ListNode) -> bool:
        # 1. Build a reverse list for the first half of the input list using slow, fast pointer
        slow, fast = head, head
        reverse_head = None

        while fast and fast.next:
            new_node = ListNode(slow.val, reverse_head)
            reverse_head = new_node
            slow = slow.next
            fast = fast.next.next

        # 2. Compare reverse list with the second half of the input list
        reverse = reverse_head
        if fast: # for odd-numbered list        
            slow = slow.next 

        while slow and reverse:
            if slow.val != reverse.val:
                return False
            slow = slow.next
            reverse = reverse.next

        return True
```

### Feedback

Your solution is correct and achieves O(n) time with O(n) auxiliary space. The reverse-prefix technique is valid: by the time fast reaches the end, reverse_head contains the first half in reverse order; skipping slow for odd lengths aligns the comparison correctly. Returning True after the comparison is safe because both halves have matching lengths. 

For an interview, explicitly mention that the extra ListNode objects make space O(n), unlike an in-place reversal approach, and note that the original list remains unchanged—a useful advantage. 

One minor readability improvement would be to use a name like reversed_first_half instead of reverse, since it represents a list rather than an operation.

```python
from typing import Callable
class Test:  
    def __init__(self, nodes: list[int], result: bool):
        self.input = nodes
        self.expected_result = result
        
        current = ListNode()
        self.head = current
        
        for i in range(len(nodes)):
            current.next = ListNode(nodes[i])
            current = current.next

        self.head = self.head.next

def run_tests(tests: list[Test], func: Callable[[ListNode], bool]):
    for i, test in enumerate(tests):
        result = func(test.head)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test([], True),
    Test([4], True),
    Test([2,2], True),
    Test([1,2], False),
    Test([1,2,2], False),
    Test([1,2,1], True),
    Test([1,2,2,1], True),
    Test([1,2,3,1], False),
    Test([5,3,1,3,5], True),
    Test([5,3,1,1,3,5], True),
    Test([5,3,1,3], False)
 ]

run_tests(tests, Solution().isPalindrome)
```
