### DESCRIPTION (inspired by Leetcode.com)
Given the root of a binary tree and an integer target, write a recursive function to determine if the tree has a root-to-leaf path where all the values along that path sum to the target.

Example 1:

Input:

[4, 2, 7, 1, 3, 6, 9]
target = 17 

Output: true (the path is 4 -> 7 -> 6)

Example 2:

Input:

[4, 2, 7, 1, 3, 6, 9]
target = 13

Output: false

```python
class TreeNode:
    def __init__(self, val: int, left: 'TreeNode' = None, right: 'TreeNode' = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def pathSum(self, root: TreeNode, target: int) -> bool:
        return self._subtreeSum(root, 0, target)

    def _subtreeSum(self, root: TreeNode, parent_sum: int, target: int) -> bool:
        if root is None:
            return False
        
        elif root.left is None and root.right is None:
            return target == (root.val + parent_sum)
        
        else:
            return (self._subtreeSum(root.left, root.val + parent_sum, target) or 
                self._subtreeSum(root.right, root.val + parent_sum, target))
```

### Feedback
Your solution is correct and handles key edge cases: an empty tree returns False, and only root-to-leaf paths are accepted. Accumulating parent_sum is a valid alternative to decrementing the target. The short-circuiting or also avoids unnecessary traversal once a path is found. 

Complexity is O(n) time and O(h) recursion stack, where h is tree height. 

For interview readability, you could simplify the branching by checking root is None first and recursively passing target - root.val; however, that is stylistic rather than a correctness improvement. 

Be prepared to explain why checking the sum only at leaves is essential.

```python
class TreeNode:
    def __init__(self, val: int, left: 'TreeNode' = None, right: 'TreeNode' = None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def pathSum(self, root: TreeNode, target: int) -> bool:
        if root is None:
            return False
        
        elif root.left is None and root.right is None:
            return target == root.val
        
        else:
            return (self.pathSum(root.left, target-root.val) or 
                self.pathSum(root.right, target-root.val))
```

```python
from typing import Callable

class Input:
    def __init__(self, nums: list[int], target: int):
        self.nums = nums
        self.target = target

class Test:  
    def __init__(self, input: Input, result: int):
        self.input = input
        self.expected_result = result
        self.root = self._createNodefromList(input.nums, 0)

    def _createNodefromList(self, nums: list[int], index: int) -> TreeNode:
        if index >= len(nums) or nums[index] is None:
            return None
        newNode = TreeNode(nums[index])
        newNode.left = self._createNodefromList(nums, 2*index+1)
        newNode.right = self._createNodefromList(nums, 2*index+2)
        return newNode

def run_tests(tests: list[Test], func: Callable[[TreeNode, int], int]):
    for test in tests:
        result = func(test.root, test.input.target)
        if result == test.expected_result:
            print(f"Test passed for {test.input.nums}")
        else:
            print(f"Test failed for {test.input.nums}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test(Input([],3),False),
    Test(Input([],0),False),
    Test(Input([0],0),True),
    Test(Input([3],3),True),
    Test(Input([5],3),False),
    Test(Input([1,2],1),False), # must from root to leaf, and root is not a left
    Test(Input([-1,3],2),True),
    Test(Input([10,-8],2),True),
    Test(Input([4, 2, 7, 1, 3, 6, 9],17),True),
    Test(Input([4, 2, 7, 1, 3, 6, 9],13),False),
]

run_tests(tests, Solution().pathSum)
```
