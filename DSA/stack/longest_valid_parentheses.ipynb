Given a string containing just the characters '(' and ')', find the length of the longest valid (well-formed) parentheses substring. A well-formed parentheses string is one that follows these rules:

Open brackets must be closed by a matching pair in the correct order.
For example, given the string "(()", the longest valid parentheses substring is "()", which has a length of 2. Another example is the string ")()())", where the longest valid parentheses substring is "()()", which has a length of 4.

Example 1:

Inputs:
s = "())))"

Output: 2

(Explanation: The longest valid parentheses substring is "()")

Example 2:

Inputs:
s = "((()()())"

Output: 8

(Explanation: The longest valid parentheses substring is "(()()())" with a length of 8)

Example 3:

Inputs:
s = ""

Output: 0


```python
class Solution:
    def longest_valid_parentheses(self, s: str) -> int:
        max_, current = 0, 0
        open = 0

        for char in s:
            if char == '(':
                open += 1
            elif char == ')':
                if open > 0:
                    open -= 1
                    current += 2
                else:
                    # no matching open parenthesis, reset counter
                    max_ = max(max_, current)
                    current = 0

        max_ = max(max_, current)
        return max_   
```

### Feedback

Your scan correctly counts matched pairs, but current assumes every match belongs to one contiguous valid substring. In ()((), after matching (), the later ( remains unmatched; the final ) matches that later (, so your code reports 4 even though the earlier () cannot connect to it. You need to track boundaries (or lengths) so unmatched opens separate valid regions. A standard O(n) approach uses a stack of indices initialized with [-1]: push ( indices; on ), pop, reset the boundary when the stack empties, otherwise update i - stack[-1]. This handles nested and adjacent substrings correctly. Also, avoid naming a variable open, since it shadows Python’s built-in open() function.



```python
class Solution:
    def longest_valid_parentheses(self, s: str) -> int:
        max_ = 0
        stack = [-1] # to keep track of all valid starting points (position of open parenthesis). Start with -1 as start of string.

        for i, c in enumerate(s):
            if c == '(':
                stack.append(i)
            else:
                stack.pop()
                if not stack:
                    # when stack is empty, it means current closing parenthesis has no matching open parenthesis, restart counting at the current position
                    stack.append(i)
                else:
                    # count the length from here to the last valid position
                    max_ = max(max_, i-stack[-1])
        
        return max_   
```

```python
from typing import Callable

class Test:  
    def __init__(self, input: str, result: int):
        self.input = input
        self.expected_result = result
        
def run_tests(tests: list[Test], func: Callable[[str], int]):
    for test in tests:
        result = func(test.input)
        if result == test.expected_result:
            print(f"Test passed for {test.input}")
        else:
            print(f"Test failed for {test.input}. Expected: {test.expected_result}, Actual: {result}")
```

```python
tests = [
    Test("())))", 2),
    Test("((()()())", 8),
    Test("", 0),
    Test("(", 0),
    Test(")", 0),
    Test("()", 2),
    Test(")()())", 4),
    Test("())(()()", 4),
    Test("()(()", 2),
    Test("()())()", 4)
]

run_tests(tests, Solution().longest_valid_parentheses)
```
